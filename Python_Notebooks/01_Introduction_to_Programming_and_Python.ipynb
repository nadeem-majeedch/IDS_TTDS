## 1. What is programming?

A **computer** is a very fast but very literal machine. It can add millions
of numbers in a blink, but it cannot "figure out" what you *mean* — you must
describe every step exactly.

**Programming** is the craft of writing those step-by-step descriptions in a
language a computer can execute. A person who writes them is a
**programmer** (or developer). The activity itself is called
**programming** or **coding**.

Think of programming as writing a recipe:

* A recipe lists ingredients and steps; a program lists **data** and
  **instructions**.
* A recipe must be followed in order; so is a program.
* A recipe that says "add some sugar" fails; a program must be told
  *exactly* how much, of what, and when.

The difference: a cook can improvise. A computer never does — it does
**exactly** what you wrote, not what you *wanted*.

## 2. What is a program? What is an algorithm?

### Program

A **program** is a sequence of instructions, written in a programming
language, that a computer executes to perform a specific task.

### Algorithm

An **algorithm** is the idea *behind* a program: a finite, precise sequence
of steps that solves a problem — written in plain language first, code later.

**Example algorithm — making tea:**

1. Boil water.
2. Put a tea bag in a cup.
3. Pour the boiling water into the cup.
4. Wait 3 minutes.
5. Remove the tea bag.

Notice the properties of a good algorithm:

* **Finite** — it ends after a limited number of steps.
* **Precise** — every step is unambiguous ("wait 3 minutes", not "wait a bit").
* **Ordered** — steps happen in sequence; you cannot pour water you have
  not boiled.
* **Effective** — each step can actually be carried out.

### A first mathematical algorithm: finding the average of three numbers

1. Get the three numbers, call them `a`, `b`, `c`.
2. Add them: `total = a + b + c`.
3. Divide the total by 3: `average = total / 3`.
4. Report `average`.

The same algorithm becomes a *program* in any language. Here it is in
Python — run the cell (Ctrl+Enter or the ▶ button):

````python
!python -m pip --version
````

````python
# A tiny program that averages three numbers.
number_a = 7        # first number
number_b = 4        # second number
number_c = 10       # third number

total = number_a + number_b + number_c   # step 2: add them
average = total / 3                      # step 3: divide by three

print("Total:", total)         # step 4a: report the total
print("Average:", average)     # step 4b: report the average
````

### Explanation

Line by line:

* `number_a = 7` — store the value 7 under the name `number_a` (a *variable*;
  fully explained in Module 2).
* `total = number_a + number_b + number_c` — compute the sum and store it.
* `average = total / 3` — compute the average and store it.
* `print(...)` — display results on the screen.
* Lines starting with `#` are **comments**: notes for humans, ignored by
  Python.

Change `number_a` to `10` in the cell above, run it again, and watch the
output change. This edit–run–observe cycle is the heart of programming.

## 3. Problem solving using computers

Real programming is mostly **problem solving**. A reliable approach:

1. **Understand** the problem. What goes in? What should come out?
2. **Plan** the algorithm in plain words or a diagram.
3. **Translate** the plan into code, a small piece at a time.
4. **Test** with known examples.
5. **Fix** mistakes (called *bugs*) and test again. This is *debugging*.

**Worked example** — Problem: "Convert a length in kilometres to miles."

1. *Understand:* input = kilometres; output = miles; 1 km ≈ 0.621371 miles.
2. *Plan:* read kilometres → multiply by 0.621371 → show result.
3. *Translate:* see the next cell.
4. *Test:* 10 km should give ≈ 6.21371 miles.
5. *Fix:* compare against a known value and correct if needed.

````python
# Problem: convert kilometres to miles.
# Plan:    miles = kilometres * 0.621371

kilometres = 10                    # input: the distance in kilometres
conversion_factor = 0.621371       # 1 kilometre in miles

miles = kilometres * conversion_factor   # the conversion step

print(kilometres, "kilometres is", miles, "miles")
````

## 4. Programming languages — high-level vs low-level

Computers only truly understand **machine code**: long strings of numbers
(in binary) that are different for each processor family. Nobody wants to
write programs that way, so we use programming languages at different
"levels":

| | Low-level languages | High-level languages |
|---|---|---|
| Examples | Machine code, Assembly | Python, Java, C++, JavaScript |
| Closeness to hardware | Very close | Far from hardware |
| Readability | Very hard for humans | Designed for humans |
| Speed of execution | Very fast | Usually a bit slower |
| Portability | Tied to one processor type | Runs on many systems |
| Lines for the same task | Many | Few |

**Assembly example** (x86, printing nothing useful to us here, just to
illustrate) — the same "add two numbers" idea:

```asm
mov eax, 5
add eax, 3
```

**Python equivalent:**

```python
result = 5 + 3
```

High-level languages are **translated** into machine code in one of two ways:

* **Compiled** (C, C++, Rust): a *compiler* translates the whole program to
  a machine-code file once; you then run that file. Fast execution, one
  extra build step.
* **Interpreted** (Python, JavaScript): an *interpreter* reads and runs your
  source code directly, statement by statement. Instant feedback — great for
  learning. Python actually compiles your code to an intermediate form
  (called *bytecode*) behind the scenes, but you never manage that yourself.

Python is a **high-level, general-purpose, interpreted** language.

## 6. The Python interpreter, source code, and execution

* **Source code** — the text you write: instructions in a programming
  language, stored in a file such as `hello.py`. It is just text.
* **Interpreter** — the program that reads source code and *executes* it.
  When you "run Python", you are really running the interpreter with your
  file as its input.

### How Python executes a program

```text
hello.py  ──▶  Python interpreter  ──▶  bytecode  ──▶  Python Virtual Machine (PVM)
(source code)   (reads & checks)        (compact        (executes the bytecode
                                         instructions)   instruction by instruction)
```

1. The interpreter **parses** your source code, checking the syntax.
2. It **compiles** it to bytecode (a compact internal instruction format).
3. The **Python Virtual Machine (PVM)** executes the bytecode, statement by
   statement, top to bottom.

If step 1 finds a **syntax error** (e.g. a missing bracket), the program
stops immediately with an error message and *nothing* runs. If the syntax is
fine but an instruction cannot be carried out (e.g. dividing by zero),
execution stops at that point with a **runtime error** — you will meet both
kinds soon.

## 9. Jupyter notebooks in practice

### Cells

A notebook is a sequence of **cells**:

  lists, tables, even maths.

* `Ctrl+Enter` — run and stay.

### The kernel

The **kernel** is the running Python process behind the notebook. It holds
every variable you create in memory. Two crucial facts:

1. **Execution order matters, not cell position.** If you run cell 10 before
   cell 3, the variables are set in the order you *ran* them. The numbers
   next to each code cell (`In [n]`) show the real order.
2. **Restart wipes memory.** "Kernel → Restart" clears every variable.
   *Restart & Run All* re-runs the whole notebook from the top — the honest
   test that a notebook works from scratch (it is how these notebooks were
   validated).

If your notebook behaves strangely (mysterious "name is not defined"
errors), Restart & Run All is the first fix to try.

## 10. Comments, indentation, syntax, keywords, identifiers

### 10.1 Comments

A **comment** is a note for human readers; Python ignores it.

* Everything after `#` on a line is a comment (**single-line comment**).
* Comments may sit on their own line or at the end of a code line.

````python
# This whole line is a comment.
print("Comments are ignored")  # This comment explains the line.
````

* For **multi-line documentation** Python uses a **docstring**: a string
  literal on its own, written with triple quotes. (Later in the course you
  will attach docstrings to functions; here they simply show the convention.)

````python
"""
This is a docstring — a multi-line comment convention.
Real programs use these to document files and, later, functions.
Running this cell prints nothing, because the text is just a value
that is never used.
"""
print("The docstring above did not print anything.")
````

*Why comments matter:* good comments explain **why** the code does
something, not merely *what* it does — the code itself shows the what.

### 10.2 Indentation

In many languages blocks of code are wrapped in `{ }`. In Python the
**indentation itself** (usually 4 spaces) marks a block:

```python
if temperature > 30:      # a condition (Module 5 explains `if`)
    print("It is hot")    # indented 4 spaces: belongs to the `if`
print("Always printed")   # not indented: outside the `if`
```

Rules:

* Use **4 spaces** per level (the PEP 8 style guide recommendation). Never
  mix tabs and spaces.
* All lines of the same block must be indented equally.
* Indentation mistakes are a *syntax error* — Python refuses to run:

```python
print("oops")
    print("extra indent")   # IndentationError: unexpected indent
```

You will use indentation for `if` (Module 5) and loops (Module 6) — both
are introduced gently when you get there.

### 10.3 Python syntax

* One **statement** per line is the norm; no `;` needed at the end.
* Text must be inside quotes: `"like this"` or `'like this'`.
* Case-sensitive: `Print` ≠ `print`.
* Groups like parentheses `()` must be balanced.
* Extra spaces are usually fine; *indentation* is the meaningful spacing.

When Python cannot understand a line, it raises a **SyntaxError** and shows
the offending spot. Every programmer sees these daily — read the message;
it usually points near the real problem.

### 10.4 Keywords

**Keywords** are reserved words with special meaning in Python. They cannot
be used as names. The cell below lists them (it uses `keyword`, a module
from the standard library — modules are Python's way of bundling extra
tools, introduced properly in later modules).

````python
import keyword

print("Python has", len(keyword.kwlist), "keywords:")
print(keyword.kwlist)
````

Notice `False`, `True`, `None`, `if`, `else`, `for`, `while` — words you
will meet throughout this course. You could never name a variable `if`,
because Python would not know whether you meant the keyword.

### 10.5 Identifiers and naming conventions

An **identifier** is a name you invent: for variables, and later for
functions and more. **Naming rules** (Python enforces these):

1. Letters, digits, and the underscore `_` only — no spaces or `-`.
2. Cannot **start** with a digit.
3. Cannot be a **keyword**.
4. Case-sensitive: `age`, `Age` and `AGE` are three different names.

**Valid:** `age`, `student_name`, `_private`, `x2`, `TOTAL`

**Invalid:** `2x` (starts with a digit), `my-var` (hyphen), `total marks`
(space), `class` (keyword)

**Naming conventions** (community style rules from [PEP 8](https://peps.python.org/pep-0008/) —
Python will not force them, but every Python programmer follows them):

* Variables: `lowercase_with_underscores` → `student_name`, `total_price`
* Constants (fixed values): `UPPERCASE` → `MAX_SPEED = 120`
* Names should be descriptive: `price` beats `p`; `marks_obtained` beats `x`.
* Avoid Python's own names (`print`, `list`, `str`) as your variable names.

````python
# Good names vs poor names — same program, very different readability.

# Poor style:
a = 12.5
b = 6
c = a * b
print(c)

# Good style:
hourly_rate = 12.5
hours_worked = 6
total_pay = hourly_rate * hours_worked
print(total_pay)
````

### 10.6 Basic program structure

A typical simple Python program flows like this:

```python
# 1. (optional) docstring or description comment at the top
"""Describe what the program does."""

# 2. (optional) imports — bring in tools you need
import keyword

# 3. (optional) constants — fixed values, UPPER_CASE
TAX_RATE = 0.17

# 4. the program's steps: statements in order
price = 100
tax = price * TAX_RATE
print("Total with tax:", price + tax)
```

Not every program needs all the parts, but this is the shape you will grow
into. Run the equivalent live version:

````python
"""Small demonstration of basic program structure."""

# Constants use UPPER_CASE names by convention.
TAX_RATE = 0.17

# Ordinary steps of the program, from top to bottom.
price = 100
tax = price * TAX_RATE
total_with_tax = price + tax

print("Price:", price)
print("Tax:", tax)
print("Total with tax:", total_with_tax)
````
