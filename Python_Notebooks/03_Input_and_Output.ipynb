## 1. Output — the `print()` function

### Concept

`print()` displays values on the screen (in a notebook: under the cell; in a
terminal: in the console). Its general shape:

```python
print(value1, value2, ..., sep=" ", end="\n")
```

* Any number of values, separated by commas — Python prints them all,
  with `sep` between them and `end` after the last.
* Default `sep` is one space; default `end` is a newline (`\n`), which is
  why each `print` starts a new line.

````python
# Printing a string:
print("Hello, students!")

# Printing a number (no quotes needed):
print(42)
print(3.75)
````

````python
# Printing multiple values with one call — separated by spaces automatically:
course_name = "Programming Fundamentals"
total_weeks = 6
print("Course:", course_name, "| Weeks:", total_weeks)
````

### The `sep` parameter

`sep` replaces the default space between values:

````python
print("2026", "09", "11", sep="-")        # date style
print("0345", "9123456", "7", sep="")     # glued together (phone style)
print("a", "b", "c", sep=" -> ")          # arrow chain
````

### The `end` parameter

`end` replaces the newline at the end. Use it to keep printing on the same
line:

````python
print("Counting:", end=" ")
print(1, end=" ")
print(2, end=" ")
print(3)
````

````python
# A small progress bar effect — three prints, one line:
print("Loading", end="")
print(".", end="")
print(".", end="")
print(".")
````

## 2. Escape characters

Some characters cannot be typed directly inside a string (a newline, a
quote that matches the string's own quotes, ...). An **escape sequence**
is a backslash + letter that stands for such a character:

| Sequence | Meaning |
|---|---|
| `\n` | newline |
| `\t` | tab |
| `\\` | a single backslash |
| `\"` | a double quote (inside `"..."`) |
| `\'` | a single quote (inside `'...'`) |

````python
print("Line one.\nLine two.")            # \n starts a new line
print("Name:\tAmina")                    # \t jumps to the next tab stop
print("She said \"wow\" loudly.")        # quotes inside double quotes
print('It\'s easy.')                     # apostrophe inside single quotes
print("A path: C:\\Users\\student")      # printing an actual backslash
````

**Formatting output** usually means mixing `\n` / `\t` with the f-strings
you will meet in section 5 — e.g. aligned table rows
`f"{name}\t{marks}"`.

## 3. Input — the `input()` function

### Concept

`input()` pauses the program, shows a prompt, and waits for the user to
type something and press Enter:

### Syntax

```python
variable = input("Prompt text: ")
```

### ⚠️ The golden rule

> **`input()` ALWAYS returns a string** — even if the user types digits.

If you need a number, you must **convert** it with `int()` or `float()`.

### Example 1 — Beginner

````python
# In this notebook, input() reads from the cell's input box.
# When you run this cell, type your name and press Enter.

name = input("Enter your name: ")
print("Hello,", name)
````

In **Jupyter**: an input box appears above the cell (or at the top of the
page) — type there and press Enter. In a **terminal** the prompt appears in
the console. Same function, same string result.

### Example 2 — Practical: the two-step conversion

Convert in a **separate step** — it is easier to read and to debug:

````python
age_text = input("Enter your age: ")   # step 1: read text
age = int(age_text)                    # step 2: convert to int

next_year_age = age + 1
print("Next year you will be", next_year_age)
````

The one-line version does exactly the same:

```python
age = int(input("Enter your age: "))
```

Read it from the inside out: first `input()` gets text, then `int()`
converts it.

### Converting different kinds of input

````python
# String input — no conversion needed:
city = input("Which city do you live in? ")
print("You live in", city)

# Integer input:
roll_number = int(input("Enter your roll number: "))
print("Roll number next year:", roll_number + 1)

# Float input:
height = float(input("Enter your height in metres: "))
print("Height in centimetres:", height * 100)
````

### Boolean-style input handling

There is no "boolean input". The standard pattern: ask for a specific word
and **compare** the (cleaned) answer — giving a True/False value:

````python
answer = input("Are you a student? (yes/no): ")

is_student = answer == "yes"     # comparison produces a bool!
print("is_student =", is_student)
````

Real users type `YES`, `Yes`, `" yes "` ... Module 4's string methods
(`lower()`, `strip()`) will make this comparison robust:

```python
is_student = answer.strip().lower() == "yes"
```

### Handling invalid input — the basic idea

What if the user types `twenty` where a number is expected?
`int("twenty")` raises a **ValueError** and the program stops.

The beginner-friendly first line of defence is to **check before
converting**. We use the string method `str.isdigit()`, which answers
"does the text contain only the digits 0–9?" (more string methods come in
Module 4):

````python
age_text = input("Enter your age: ")

if age_text.isdigit():                    # only digits 0-9 in the text?
    age = int(age_text)
    print("In 5 years you will be", age + 5)
else:
    print("That is not a whole number — please run again.")
````

A second technique: the **conversion with a safety net**. `try`/`except`
catches the error gracefully. It is shown here for completeness — proper
error handling is a later-course topic:

````python
text = input("Enter a number: ")

try:
    number = float(text)
    print("Twice your number is", number * 2)
except ValueError:
    print("Invalid number — using 0 instead.")
    number = 0.0
````

### Common Mistakes

```python
# 1. Forgetting the string rule:
age = input("Age: ")
# print(age + 1)              # TypeError: str + int

# 2. Converting with str() when you meant int():
total = int("5") + int("7")   # 12 — correct
total_wrong = str(5) + str(7) # "57" — glued text!

# 3. Empty prompt:
input()                       # user sees a blinking cursor, no clue what to type

# 4. int() on decimal text:
# int("9.5")                  # ValueError — use float("9.5") instead
```

## 4. f-strings — formatting output properly

### Concept

An **f-string** is a string with an `f` before the opening quote. Inside it,
`{...}` holds any expression — Python inserts its value into the text:

### Syntax

```python
f"text {variable} more text {expression} text"
```

````python
name = "Amina"
age = 19

print(f"Hello {name}!")
print(f"Next year {name} will be {age + 1}.")
````

f-strings beat `+`-concatenation: no `str()` calls, no spacing mistakes,
and any expression works inside `{...}`.

````python
# The old way vs f-strings:
price = 249.5
quantity = 3

old_way = "Total: " + str(price * quantity) + " PKR"
new_way = f"Total: {price * quantity} PKR"

print(old_way)
print(new_way)
````

### Formatting numbers: decimals, padding, percent

Inside the braces, a `:` starts a **format spec**. The workhorse is
`.Nf` — *N decimal places, fixed-point*:

````python
pi = 3.141592653589793

print(f"pi = {pi:.2f}")         # 2 decimal places
print(f"pi = {pi:.4f}")         # 4 decimal places
````

### More formatting in action

The same idea with shopping-style values — decimals, percent, and fixed
widths for aligned columns:

````python
pi = 3.141592653589793
total_bill = 4550.0
discount_fraction = 0.15

print(f"pi = {pi:.2f}")                       # 2 decimals
print(f"pi = {pi:.4f}")                       # 4 decimals
print(f"Bill: {total_bill:.1f} PKR")          # 1 decimal
print(f"Discount: {discount_fraction:.0%}")   # percent style -> 15%
print(f"Wide: [{total_bill:10.2f}]")          # width 10, right-aligned
print(f"Left: [{total_bill:<10.2f}]")         # width 10, left-aligned
````

Spec cheat-sheet:

| Spec | Meaning | Example → Output |
|---|---|---|
| `:.2f` | 2 fixed decimals | `f"{2.71828:.2f}"` → `2.72` |
| `:8.2f` | width 8 + 2 decimals | `f"{45.5:8.2f}"` → `   45.50` |
| `:.0%` | percentage | `f"{0.256:.0%}"` → `26%` |
| `:,` | thousands separator | `f"{2450000:,}"` → `2,450,000` |

### Common Mistakes

```python
# 1. Forgetting the f prefix — braces print literally:
name = "Ali"
print("Hello {name}")     # Hello {name}  (oops)
print(f"Hello {name}")    # Hello Ali     (correct)

# 2. Quotes inside braces:
# f"{'yes' if 5 > 3 else 'no'}" — legal, but confusing at this stage;
# compute first, then insert.
```

## 5. Complete interactive programs

### 5.1 Personal introduction program

````python
# ----- Personal Introduction -----
# Sample run:  name=Amina, city=Lahore, age=19, favourite subject=Statistics

name = input("What is your name? ")
city = input("Which city are you from? ")
age = int(input("How old are you? "))
favourite_subject = input("Your favourite subject? ")

print()
print("===== Introduction Card =====")
print(f"Name:      {name}")
print(f"City:      {city}")
print(f"Age:       {age} years old")
print(f"Favourite: {favourite_subject}")
print("================================")
````

**Explanation:** four `input()` calls collect text; only `age` needs
conversion (it is used in a sentence, so it could even stay text — but we
convert because a future version will do maths with it). `print()` with no
argument prints an empty line for spacing.

### 5.2 Simple calculator (two numbers, four operations)

````python
# ----- Simple Calculator -----
# Sample run:  12 and 5

first_number = float(input("Enter the first number: "))
second_number = float(input("Enter the second number: "))

print()
print(f"{first_number} + {second_number} = {first_number + second_number}")
print(f"{first_number} - {second_number} = {first_number - second_number}")
print(f"{first_number} * {second_number} = {first_number * second_number}")
print(f"{first_number} / {second_number} = {first_number / second_number}")
````

### 5.3 Student information program

````python
# ----- Student Information -----
# Sample run:  Bilal, CS-101, 18, 3, 72.5

student_name = input("Student name: ")
roll_number = input("Roll number: ")
age = int(input("Age: "))
semester = int(input("Current semester: "))
lab_marks = float(input("Lab marks (out of 100): "))

print()
print(f"Student : {student_name} ({roll_number})")
print(f"Age     : {age} years, semester {semester}")
print(f"Lab avg : {lab_marks:.1f}/100")
````

### 5.4 Marks/percentage calculator

````python
# ----- Percentage Calculator (5 subjects) -----
# Sample run:  78, 84, 91, 65, 88 — each out of 100

marks_one = float(input("Marks in subject 1: "))
marks_two = float(input("Marks in subject 2: "))
marks_three = float(input("Marks in subject 3: "))
marks_four = float(input("Marks in subject 4: "))
marks_five = float(input("Marks in subject 5: "))

total_marks = marks_one + marks_two + marks_three + marks_four + marks_five
maximum_marks = 500.0
percentage = total_marks / maximum_marks * 100

print()
print(f"Total obtained : {total_marks:.0f}/{maximum_marks:.0f}")
print(f"Percentage     : {percentage:.2f}%")
````

### 5.5 Temperature converter

````python
# ----- Temperature Converter -----
# Sample run:  37

celsius = float(input("Temperature in °C: "))
fahrenheit = celsius * 9 / 5 + 32

print(f"{celsius:.1f} °C  =  {fahrenheit:.1f} °F")
````

### 5.6 Shopping bill calculator

````python
# ----- Shopping Bill -----
# Sample run:  item=Rice bag, price=1850.0, qty=2, discount%=10

item_name = input("Item name: ")
unit_price = float(input("Price per unit (PKR): "))
quantity = int(input("Quantity: "))
discount_percent = float(input("Discount percent: "))

subtotal = unit_price * quantity
discount_amount = subtotal * discount_percent / 100
final_bill = subtotal - discount_amount

print()
print("========== RECEIPT ==========")
print(f"Item          : {item_name}")
print(f"Unit price    : {unit_price:.2f} PKR")
print(f"Quantity      : {quantity}")
print(f"Subtotal      : {subtotal:.2f} PKR")
print(f"Discount      : -{discount_amount:.2f} PKR")
print("-----------------------------")
print(f"FINAL BILL    : {final_bill:.2f} PKR")
print("=============================")
````
