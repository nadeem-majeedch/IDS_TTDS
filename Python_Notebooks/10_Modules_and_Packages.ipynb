## 1. What is a module?

### Concept

A **module** is a plain Python file (name ending in `.py`) containing
variables, functions — anything. **Importing** a module makes its names
available in your program under the module's name.

### Why do we need it?

* **Organisation** — thousands of lines in one file is unreadable.
* **Reuse** — write `math_utils.py` once, import it in every project.
* **Separation** — display code, calculation code and storage code live in
  different files and change for different reasons.
* **The standard library** — decades of tested tools (`math`, `random`,
  `csv`, `json`, ...) come with Python. Notebook 00 installed even *more*
  via pip (Module 0's environment skills).

The analogy: a book (your program) is easier to use with **chapters**
(modules) than as one unbroken wall of text.

## 2. The three import forms

### Syntax

```python
import math                    # form 1: the whole module, names via math.x
from math import sqrt          # form 2: specific names, no prefix
import math as m               # form 3: the module under an alias
```

````python
import math

print(math.sqrt(16))        # module name . function
print(math.pi)              # constants too
````

````python
from math import sqrt, pi

print(sqrt(25))             # used directly — no prefix
print(round(pi, 2))
````

````python
import math as m

print(m.floor(7.9))         # m is the same module, shorter name
````

### Which form when?

* `import module` — the default; the `math.` prefix documents where names
  come from (recommended for beginners).
* `from module import name` — convenient for a few heavily used names.
* `import module as alias` — conventional aliases only, e.g.
  `import statistics as stats` (you will meet `import numpy as np` in data
  science courses — same idea).

## 3. Namespaces at a beginner level

A **namespace** is the mapping between names and objects that Python
keeps. Each module has its own, so `math.log` and *your* `log` can coexist
without collision:

````python
import math

def log(message):                       # our own log — no clash with math.log
    print("LOG:", message)


log("started")
print(math.log(math.e))                 # natural log of e == 1.0
````

The prefix `math.log` reaches into the *module's* namespace; the bare
`log` stays in *yours*. That separation is exactly why the
`import module` form is the safest habit.

## 4. Standard library tour — `math`

### Example 1 — Beginner: constants and rounding

````python
import math

print(math.pi)          # circle constant
print(math.e)           # Euler's number
print(math.floor(4.7))  # round DOWN to integer
print(math.ceil(4.2))   # round UP to integer
print(math.fabs(-8))    # absolute value as a float
````

### Example 2 — Practical: powers, roots and geometry

````python
import math

side = 3.0
diagonal = math.sqrt(side ** 2 + side ** 2)     # Pythagoras
print(f"Square diagonal: {diagonal:.3f}")

radius = 5
area = math.pi * radius ** 2
print(f"Circle area: {area:.2f}")

print(math.gcd(48, 60))         # greatest common divisor
print(math.factorial(5))        # 5! — the loop-free way
````

(Module 9's recursive `factorial` now has a battle-tested twin.)

## 5. Standard library tour — `random`

### Concept

`random` produces pseudo-random numbers — deterministic sequences that
*look* random, seeded from the clock. For teaching (and testing!), we fix
the seed first: **the same seed produces the same sequence**, so the
outputs below are reproducible.

### Example 1 — Beginner: randint and choice

````python
import random

random.seed(42)                 # reproducible runs — remove for real randomness

print(random.randint(1, 6))     # one die roll (both ends included)
print(random.randint(1, 6))

colours = ["red", "green", "blue"]
print(random.choice(colours))   # one random element
````

### Example 2 — Practical: shuffle and sampling

````python
import random

random.seed(42)

deck = ["A", "2", "3", "4", "5"]
random.shuffle(deck)            # rearrange IN PLACE, like lists' sort()
print("Shuffled deck:", deck)

lottery = random.sample(deck, 3)   # three DIFFERENT elements, no repeats
print("Lottery draw :", lottery)
````

(`shuffle` mutates like `list.sort()`; `sample` leaves the original alone
— the in-place vs new-value distinction from Module 7 again!)

## 6. Standard library tour — `datetime`

### Example 1 — Beginner: now, today, and pieces

````python
from datetime import datetime, date

now = datetime.now()
print(now.year, now.month, now.day)     # pieces of the current moment

today = date.today()
print(today)                            # ISO format: YYYY-MM-DD
````

(Your year/month/day values will, correctly, differ from any frozen
output — that is the point of "now".)

### Example 2 — Practical: creating dates and date arithmetic

````python
from datetime import date, timedelta

exam_day = date(2026, 12, 15)          # a date we choose
today = date(2026, 9, 12)              # stand-in for date.today()

days_left = exam_day - today           # subtracting dates -> timedelta
print(days_left)
print("Weeks left:", days_left.days // 7)

deadline = today + timedelta(days=30)  # add time to a date
print("Deadline:", deadline)
````

## 7. Standard library tour — `statistics`

### Example 1 — Beginner: the three averages

````python
import statistics

marks = [78, 91, 65, 91, 84]

print(statistics.mean(marks))     # the average
print(statistics.median(marks))   # middle value of the sorted list
print(statistics.mode(marks))     # most frequent value
````

### Example 2 — Practical: spread of the data

````python
import statistics

class_a = [70, 72, 71, 69, 73]      # consistent students
class_b = [40, 95, 55, 88, 60]      # wildly different students

print("A: mean", statistics.mean(class_a), "stdev", round(statistics.stdev(class_a), 2))
print("B: mean", statistics.mean(class_b), "stdev", round(statistics.stdev(class_b), 2))
````

Same-ish averages, wildly different spread — the **standard deviation**
tells them apart. Data science lives on ideas like this; here it is in one
standard-library call.

### Example 1 — Beginner: import the whole module

### How Python finds modules (beginner version)

When you `import math_utils`, Python searches:

1. the **current folder** (for scripts: the script's folder; for Jupyter:
   the notebook's folder),
2. the folders of installed packages (site-packages),
3. the standard library locations.

First hit wins — which is why `math_utils.py` living next to this notebook
just works, and why you should **not** name your file `random.py` or
`math.py` (it would shadow the real module!).

## 9. Packages and `__init__.py`

### Concept

A **package** is a folder of modules with an `__init__.py` file inside.
The `__init__.py` (often empty) marks the folder as a package and can set
up imports. Dotted paths address modules inside:
`package.module.function`.

This repository ships a tiny one:

```text
notebooks/
└── demo_package/
    ├── __init__.py
    ├── greetings.py      (say_hello, say_goodbye)
    └── text_tools.py     (shout, word_count, initials)
```

````python
from demo_package import greetings, text_tools

print(greetings.say_hello("Amina"))
print(greetings.say_goodbye("Bilal"))
print(text_tools.shout("well done"))
print(text_tools.word_count("one two three four"))
print(text_tools.initials("Alia Khan"))
````

````python
# The other import forms work on packages too:
import demo_package.text_tools as tt

print(tt.initials("Data Science Society"))

from demo_package.text_tools import shout

print(shout("modules"))
````

**Why packages?** A module is a chapter; a package is the book's part
("Chapters 5–8"). Big projects group related modules so imports stay
readable: `from university import grading`, `from university import
attendance`.

## 10. Good project structure

A beginner-friendly layout for a multi-file project:

```text
student_system/
├── main.py                <- entry point: starts the program
├── students.py            <- student data + student functions
├── grading.py             <- marks → grades logic
├── report.py              <- printing/formatting helpers
└── utils/
    ├── __init__.py
    └── dates.py           <- small shared helpers
```

Principles:

* `main.py` stays tiny — it imports and orchestrates.
* Each file has **one clear responsibility** (Module 16 returns to this).
* Shared helpers sink into utility modules/packages.
* Files import each other: `from grading import letter_grade`.

## 12. Common mistakes

```python
# 1. Naming your file random.py/math.py — shadows the standard module.
# 2. import file.py — the .py is NEVER part of the import (import file).
# 3. from math import * — floods your namespace; imports should be explicit.
# 4. Forgetting the module prefix: sqrt(16) after plain `import math`.
# 5. Side effects at import time — modules should define, not print/loop.
# 6. Moving a notebook but leaving the helper module behind -> ModuleNotFoundError.
```

## Find the Error

**Error 1**

```python
import math
print(sqrt(16))
```

**Error 2**

```python
import random.py
print(random.randint(1, 2))
```

**Error 3**

```python
from math import *
from my_math import *
result = sqrt(4)      # which sqrt is this?
```

**Error 4**

```python
from demo_package import greetings
print(say_hello("Ali"))
```

**Error 5**

```python
# my_module.py starts with:
print("my_module loaded!")
PI = 3.14
```

<details><summary>Answers</summary>

1. `NameError` — plain `import math` requires the prefix: `math.sqrt(16)`.
2. No `.py` in import statements: `import random`.
3. Star imports collide silently — the later import wins. Name your
   imports explicitly to stay in control.
4. `say_hello` lives *inside* the module: use
   `greetings.say_hello("Ali")` or
   `from demo_package.greetings import say_hello`.
5. Importing now prints as a side effect. Modules should only define
   things.

</details>

## 15. Summary
