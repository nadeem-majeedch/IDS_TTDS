## 1. Boolean logic — the foundation

Before writing any `if`, be fluent with the **booleans** it depends on.

* A boolean value is either `True` or `False` (the only two, capitalised!).
* A **boolean expression** is any expression whose value is `True` or
  `False` — usually built with **comparison operators**.

| Comparison | Meaning |
|---|---|
| `==` | equal to |
| `!=` | not equal |
| `<`, `<=` | less (or equal) |
| `>`, `>=` | greater (or equal) |

````python
print(10 > 3)          # True
print(10 == 3)         # False
print(10 != 3)         # True
marks = 65
print(marks >= 50)     # boolean expression — evaluates to True
````

### Logical operators: `and`, `or`, `not`

They combine boolean expressions:

| `a` | `b` | `a and b` | `a or b` | `not a` |
|---|---|---|---|---|
| True | True | **True** | True | False |
| True | False | False | **True** | False |
| False | True | False | **True** | True |
| False | False | False | False | True |

* `and` — True only when **both** sides are True.
* `or` — True when **at least one** side is True.
* `not` — flips the value.

````python
age = 20
has_id = True

print(age >= 18 and has_id)     # both True          -> True
print(age < 18 or has_id)       # one True is enough -> True
print(not has_id)               # flip               -> False
````

````python
# Chaining comparisons is natural in Python:
temperature = 25
print(18 <= temperature <= 30)   # between 18 and 30? True
````

## 2. The `if` statement

### Concept

Programs need to **decide**: do this *only if* something is true. The `if`
statement runs its indented block exactly when its condition is `True`;
otherwise the block is skipped and the program continues.

### Syntax

```python
if condition:
    statement          # runs when condition is True
    statement          # (any number of indented lines)
statement_after       # ALWAYS runs — not indented
```

Note the **colon `:`** ending the `if` line and the **4-space indentation**
of the block.

### Example 1 — Beginner

````python
temperature = 35

if temperature > 30:
    print("It is hot today.")
    print("Drink plenty of water.")

print("(This line is outside the if — it always runs.)")
````

### Explanation

Line by line:

* `temperature = 35` — store 35.
* `if temperature > 30:` — evaluate `35 > 30` → `True`, so the indented
  block below runs.
* The last `print` is **not indented**, so it is outside the `if` and runs
  no matter what.

Change `temperature` to `25` and run again: the two inside lines vanish,
the outside line stays.

````python
temperature = 25          # same code, new value — now the block is skipped

if temperature > 30:
    print("It is hot today.")

print("(outside — always runs)")
````

### Example 2 — Practical: a low-stock warning

````python
stock = 4

if stock < 5:
    print("WARNING: only", stock, "items left — reorder soon!")
````

## 3. `if ... else` — two paths

### Concept

Often exactly one of two things must happen. `else` catches the "not `if`"
case. The `else` has **no condition** of its own.

### Syntax

```python
if condition:
    ...            # runs when condition is True
else:
    ...            # runs when condition is False
```

### Example 1 — Beginner: even or odd

````python
number = 7

if number % 2 == 0:
    print(number, "is even")
else:
    print(number, "is odd")
````

### Explanation

* `number % 2` is the remainder after dividing by 2 — `0` for even numbers,
  `1` for odd.
* The condition `number % 2 == 0` compares that remainder with 0.
* For 7: remainder 1 → condition `False` → the `else` block runs.

### Example 2 — Practical: pass/fail

````python
marks = 42

if marks >= 50:
    print("Result: PASS")
else:
    print("Result: FAIL")
````

## 4. `if ... elif ... else` — many paths

### Concept

`elif` (short for *else if*) adds more conditions. Python checks each
condition **top to bottom** and runs the block of the **first** one that is
True — everything after is skipped. `else` (optional) catches all remaining
cases.

### Syntax

```python
if condition_one:
    ...
elif condition_two:
    ...
elif condition_three:
    ...
else:
    ...
```

**Order matters!** Conditions are tested top to bottom, so put the most
specific/strictest first (in grading: the highest grade).

### Example 1 — Beginner: grade calculator

````python
marks = 78

if marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "F"

print("Marks:", marks, "-> Grade:", grade)
````

### Explanation

For 78: `78 >= 80`? No. `78 >= 70`? **Yes** → `grade = "B"`, and the other
branches are not even checked. Why must `A` be tested first? If `>= 50`
were first, every mark ≥ 50 would get D — the first matching branch wins.

### Example 2 — Practical: electricity bill with slabs

````python
# First 100 units: 5.0/unit; next 100: 8.0/unit; above 200: 12.0/unit.
units = 250

if units <= 100:
    bill = units * 5.0
elif units <= 200:
    bill = 100 * 5.0 + (units - 100) * 8.0
else:
    bill = 100 * 5.0 + 100 * 8.0 + (units - 200) * 12.0

print(f"Units: {units}, Bill: {bill:.0f} PKR")
````

### Explanation

Each branch builds on the previous slabs: the first 100 units always cost
5 each, the next 100 cost 8 each, only units beyond 200 cost 12. For 250
units: 500 + 800 + 600 = 2100.

## 5. Indentation — the rules, carefully

Python uses indentation to define blocks. Four rules to memorise:

1. Use **4 spaces** per level (PEP 8). Configure your editor to insert
   spaces when Tab is pressed.
2. Every line in the same block must have the **same** indentation.
3. **Deeper** indent = inside the block; returning to a shallower indent
   ends the block.
4. Never mix tabs and spaces (Python 3 raises `TabError`).

```python
if marks >= 50:
    print("Pass")           # inside: 4 spaces
    print("Well done")      # inside: same 4 spaces
print("Done")               # outside: no indent — runs always
```

**Errors you will meet:**

```python
if marks >= 50:
print("Pass")        # IndentationError: expected an indented block

    print("Pass")    # IndentationError: unexpected indent (no block above)
```

Nested blocks simply indent deeper (8 spaces, then 12, ...):

````python
age = 19
has_national_id = True

if age >= 18:
    print("Adult check:")
    if has_national_id:                       # nested if: indented 4 more
        print("  Eligible to vote.")          # 8 spaces
    else:
        print("  Get your ID card first.")
````

## 6. Conditions in practice

### Conditions with numbers and strings

````python
# Numbers: compare with comparison operators.
balance = 1200
if balance >= 1000:
    print("Healthy balance")

# Strings: equality and ordering (case-sensitive — see Module 4).
password = "OpenSesame"
if password == "OpenSesame":
    print("Access granted")
````

````python
# Normalise text before comparing user-typed values:
answer = "YES"
if answer.lower() == "yes":
    print("Confirmed (case-insensitive)")
````

### Combining conditions with `and`, `or`, `not`

````python
age = 20
has_ticket = True
is_student = True

# Both must hold:
if age >= 18 and has_ticket:
    print("Entry allowed")

# At least one suffices:
if is_student or age < 12:
    print("Discount applies")

# Negation:
if not is_student:
    print("Full price")
else:
    print("Student rate booked")
````

````python
# Range checks with and / or:
height_cm = 165
if height_cm >= 150 and height_cm <= 190:
    print("Within ride height limits")

day = "Sunday"
if day == "Saturday" or day == "Sunday":
    print("It is the weekend")
````

### Membership conditions (`in`)

````python
vowels = "aeiou"
letter = "e"

if letter in vowels:
    print(letter, "is a vowel")

email = "student@university.edu"
if "@" in email:
    print("Looks like an email address")
````

### Example 3 — Challenge: nested vs combined

The same logic can often be written nested **or** combined. Compare:

````python
# Version A — nested:
age = 22
income = 80000

if age >= 18:
    if income >= 50000:
        print("A: Loan approved")

# Version B — combined (usually preferred: flatter, easier to read):
if age >= 18 and income >= 50000:
    print("B: Loan approved")
````

### Common Mistakes

```python
# 1. Using = instead of ==:
# if marks = 50:              # SyntaxError — use ==

# 2. Forgetting the colon:
# if marks >= 50              # SyntaxError — line must end with ':'

# 3. Wrong indentation:
# if marks >= 50:
# print("Pass")               # IndentationError

# 4. Unreachable branch — order your elifs from strictest to loosest:
# if marks >= 50:   grade = "D"
# elif marks >= 80: grade = "A"   # never reached! every mark >= 80 is also >= 50

# 5. Comparing with the wrong case:
# if answer == "Yes":  ...   # "yes" would fail — normalise with .lower()

# 6. Empty branch — use `pass` as a placeholder if you must:
if marks >= 90:
    pass                    # 'pass' does nothing; fills the block
print("continues normally")
```

## 7. Conditional expression (ternary)

### Concept

Once `if/else` is comfortable: a **conditional expression** chooses between
two *values* in a single line. Use it only when both branches are short.

### Syntax

```python
value_if_true if condition else value_if_false
```

### Example

````python
marks = 73

result = "Pass" if marks >= 50 else "Fail"
print(result)
````

````python
# A few realistic one-liners:
temperature = 31
weather_word = "hot" if temperature > 30 else "pleasant"

items_sold = 0
item_word = "items" if items_sold != 1 else "item"

print(f"The weather is {weather_word}.")
print(f"You sold {items_sold} {item_word}.")
````

Rule of thumb: if the one-liner needs a second look to understand, use a
normal `if/else` instead.

## 8. Practical examples, step by step

### 8.1 Positive / negative / zero

````python
number = -7

if number > 0:
    print("The number is positive")
elif number < 0:
    print("The number is negative")
else:
    print("The number is zero")
````

### 8.2 Largest of two numbers

````python
first = 34
second = 89

if first > second:
    print("Largest:", first)
else:
    print("Largest:", second)
````

### 8.3 Largest of three numbers

````python
first = 15
second = 42
third = 27

if first >= second and first >= third:
    largest = first
elif second >= first and second >= third:
    largest = second
else:
    largest = third

print("Largest:", largest)
````

**Explanation:** check whether `first` beats *both* others; else check
whether `second` beats both; otherwise `third` must be the largest.

### 8.4 Age eligibility (voting)

````python
age = 16

if age >= 18:
    print("You can vote.")
else:
    years_left = 18 - age
    print(f"You can vote in {years_left} year(s).")
````

### 8.5 Login validation (fixed example credentials)

````python
# NOTE: real systems never store plain passwords — this is a teaching example.
stored_username = "admin"
stored_password = "python123"

entered_username = "admin"
entered_password = "python2026"

if entered_username == stored_username and entered_password == stored_password:
    print("Login successful — welcome,", entered_username)
else:
    print("Login failed: wrong username or password.")
````

**Why not two separate ifs?** With separate messages ("wrong username",
"wrong password") you leak information to attackers. One combined check is
both safer and simpler.

### 8.6 Simple calculator using conditions

````python
first_number = 12.0
second_number = 4.0
operation = "/"        # try also: +, -, *, /

if operation == "+":
    print(f"{first_number} + {second_number} = {first_number + second_number}")
elif operation == "-":
    print(f"{first_number} - {second_number} = {first_number - second_number}")
elif operation == "*":
    print(f"{first_number} * {second_number} = {first_number * second_number}")
elif operation == "/":
    if second_number != 0:
        print(f"{first_number} / {second_number} = {first_number / second_number}")
    else:
        print("Cannot divide by zero!")
else:
    print("Unknown operation:", operation)
````

### 8.7 Simple menu selection

````python
print("===== LIBRARY MENU =====")
print("1. Borrow a book")
print("2. Return a book")
print("3. Check balance")

choice = "2"

if choice == "1":
    print("Book borrowed. Due in 14 days.")
elif choice == "2":
    print("Book returned. Thank you!")
elif choice == "3":
    print("Outstanding fine: 0 PKR")
else:
    print("Invalid choice — please pick 1, 2 or 3.")
````

### 8.8 Student grading system (the full picture)

````python
# Marks in five subjects, each out of 100:
maths = 82
physics = 74
chemistry = 91
english = 68
computing = 88

total = maths + physics + chemistry + english + computing
percentage = total / 500 * 100

if percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "F"

print(f"Total: {total}/500")
print(f"Percentage: {percentage:.1f}%")
print(f"Grade: {grade}")
````

### Find the Error

**Error 1**

```python
if temperature > 30
    print("hot")
```

**Error 2**

```python
if age = 18:
    print("adult")
```

**Error 3**

```python
if marks >= 80:
    grade = "A"
print(grade)
```

(When does Error 3 fail — with marks 85 or with marks 40? Why?)

**Error 4**

```python
if 80 <= marks:
    print("A")
elif marks <= 79:
    print("B")
```

<details><summary>Answers</summary>

1. Missing colon after the condition.
2. `=` assigns, `==` compares.
3. It fails with **40**: the `if` is False, `grade` is never created, and
   `print(grade)` raises NameError. With 85 it works by luck. Fix: give
   `grade` a value in an `else`, or initialise it before the `if`.
4. The logic misses middle values: a mark of 79.5 matches neither branch
   (and real marks can be fractional). Fix: invert the ranges cleanly and
   end with an `else` that catches everything remaining.

</details>
