## 1. Why functions?

### Concept

A **function** is a named, reusable block of code. Define it once, call it
many times.

### Why do we need it?

Imagine a 300-line program that computes a grade in six different places.
Without functions, the same (or slightly different!) grading code appears
six times — fix a bug in one copy and the other five stay broken. With a
function, the logic lives in exactly one place.

Benefits, in order of importance:

1. **Reuse** — write once, call anywhere.
2. **Naming** — `calculate_average(marks)` documents *what* happens.
3. **Testing** — one function is easy to check in isolation.
4. **Decomposition** — big problems become small, readable pieces (Module 13).

### Syntax

```python
def function_name(parameters):
    """docstring"""
    body
    return value
```

## 2. Defining and calling

### Example 1 — Beginner: the simplest function

````python
def greet():
    print("Hello!")


greet()          # call it — note the parentheses
greet()          # as many times as you like
````

### Explanation

* `def greet():` **defines** the function: the name is `greet`, it takes no
  inputs, and everything indented below belongs to it. Definition alone
  runs *nothing*.
* `greet()` **calls** it: Python jumps into the body, executes the print,
  and returns to the call site.

### Example 2 — Practical: a parameter makes it flexible

````python
def greet_person(name):
    print(f"Hello {name}, welcome to the course!")


greet_person("Amina")
greet_person("Bilal")
````

`name` is a **parameter** — a placeholder that receives a value. `"Amina"`
is an **argument** — the actual value passed in a call. Memorise the pair:

* **parameter** = the variable in the definition
* **argument** = the value in the call

### Example 3 — Challenge: several parameters

````python
def describe_pet(animal, name):
    print(f"I have a {animal} called {name}.")


describe_pet("cat", "Mano")
describe_pet("dog", "Rex")
````

### Common Mistakes

```python
# 1. Calling before defining (Python reads top to bottom):
# say_hi()          # NameError — the def comes later

# 2. Forgetting the parentheses in the call:
# greet             # this only *mentions* the function, runs nothing

# 3. Wrong number of arguments:
# describe_pet("cat")   # TypeError: missing 1 required positional argument
```

## 3. `print()` vs `return` — the great confusion

### Concept

* `print()` shows a value **on the screen** — for humans.
* `return` hands a value **back to the caller** — for the program.

A function that only prints computes something and *tells the world*, but
its caller receives `None` — it cannot build on the result. A function
that returns gives its caller a value to store, combine, or pass on.

### Example 1 — the difference, concretely

````python
def add_print(a, b):
    print(a + b)            # shows on screen, returns None


def add_return(a, b):
    return a + b            # gives the value back


result = add_print(2, 3)
print("add_print returned:", result)

result = add_return(2, 3)
print("add_return returned:", result)
print("Double it:", result * 2)      # the returned value is USABLE
````

Notice: `add_print(2, 3)` printed `5` *by itself*, then the assignment got
`None`. Only `add_return`'s value can be reused (`result * 2`).

### Example 2 — Practical: returning beats printing

````python
def average(marks):
    return sum(marks) / len(marks)


class_average = average([78, 91, 65])
print(f"Average: {class_average:.1f}")
print(f"Rounded: {round(class_average)}")     # reusable!
````

### Common Mistakes

```python
# 1. Using print when the caller needs the value (the add_print trap).
# 2. Writing return inside a loop meant to run for every element —
#    return exits the function IMMEDIATELY.
# 3. Expecting code after return to run — it never does.
```

## 4. Positional, keyword and default arguments

### Concept

* **Positional** — matched by order: `describe_pet("cat", "Mano")`.
* **Keyword** — matched by name, any order: `describe_pet(name="Mano", animal="cat")`.
* **Default** — a parameter gets a fallback value used when the caller omits it.

### Example 1 — Beginner: defaults

````python
def student_info(name, age=18, program="Undeclared"):
    print(f"{name}, {age}, {program}")


student_info("Ali")                          # all defaults after name
student_info("Sara", 20)                     # positional fills age
student_info("Omar", program="Data Science") # skip age, name program
student_info("Zainab", age=19, program="Statistics")
````

### Explanation

* Default parameters come **after** required ones (`name` first).
* Callers can mix positional and keyword arguments; keyword arguments must
  come after positional ones.

### Example 2 — Practical: a flexible grading function

````python
def grade_report(name, marks, pass_mark=50):
    total = sum(marks)
    percentage = total / len(marks)
    status = "PASS" if percentage >= pass_mark else "FAIL"
    return f"{name}: {percentage:.0f}% ({status})"


print(grade_report("Ali", [80, 90, 70]))
print(grade_report("Sara", [45, 55], pass_mark=60))
````

### Common Mistakes

```python
# 1. Defaults BEFORE required parameters:
# def f(age=18, name):        # SyntaxError

# 2. Positional after keyword in the call:
# student_info(name="Ali", 20)    # SyntaxError

# 3. Never use a mutable default like def f(items=[]): — a famous trap
#    (all calls share ONE list). Use None and create inside, Module 16 preview.
```

## 5. Multiple return values

### Concept

`return` can send back several values packed into a **tuple** (Module 8) —
and the caller unpacks them in one line.

````python
def min_max(numbers):
    return min(numbers), max(numbers)


scores = [42, 7, 88, 19]
lowest, highest = min_max(scores)      # tuple unpacking!

print("Lowest :", lowest)
print("Highest:", highest)
````

### Example — Practical: statistics bundle

````python
def summary(numbers):
    """Return (count, total, average) of the numbers."""
    count = len(numbers)
    total = sum(numbers)
    avg = total / count if count > 0 else 0
    return count, total, avg


count, total, avg = summary([3, 5, 7])
print(f"count={count}, total={total}, average={avg}")
````

## 6. Scope: local vs global

### Concept

* A variable created **inside** a function is **local** — it exists only
  during that call, then vanishes.
* A variable created **outside** all functions is **global** — readable
  everywhere.
* Assigning to a global *inside* a function needs the `global` keyword —
  and is usually a design smell. Passing values *in* and *returning* them
  is almost always better.

````python
course_name = "Programming Fundamentals"   # global

def show_course():
    inside_message = "local to show_course"   # local
    print(course_name)          # reading a global: fine
    print(inside_message)


show_course()

# The next line would fail — inside_message no longer exists out here:
# print(inside_message)       # NameError
````

````python
def reset_counter_wrong():
    counter = 0            # creates a NEW LOCAL — the global is untouched
    return counter


counter = 100
reset_counter_wrong()
print(counter)             # still 100!
````

`global` exists for the rare case where a function must *rebind* a global
name:

````python
attempts = 0                    # global

def register_attempt():
    global attempts             # declare intent to modify the GLOBAL
    attempts += 1


register_attempt()
register_attempt()
print("Attempts:", attempts)
````

**Why is heavy use of globals discouraged?** Any function could change a
global at any time, so a bug in one place can silently corrupt data used in
another. Functions that take parameters and return values keep data flowing
through visible channels — much easier to test and debug. Treat `global`
as a tool you should almost never need.

## 7. Flexible arguments: `*args` and `**kwargs`

### Concept

* `*args` collects **extra positional arguments** into a **tuple**.
* `**kwargs` collects **extra keyword arguments** into a **dictionary**.

The names `args`/`kwargs` are convention; the stars do the work.

````python
def total_all(*args):
    print("args received:", args, "of type", type(args).__name__)
    return sum(args)


print(total_all(1, 2))
print(total_all(10, 20, 30, 40))
````

````python
def show_profile(**kwargs):
    print("kwargs received:", kwargs, "of type", type(kwargs).__name__)
    for key, value in kwargs.items():
        print(f"  {key}: {value}")


show_profile(name="Ali", age=21, program="Data Science")
````

### Explanation

Inside `total_all`, `args` is just a **tuple** — everything you know about
tuples applies. Inside `show_profile`, `kwargs` is a **dictionary**. The
caller writes arguments normally; Python packs them for you.

### Example — Practical: a logger that accepts anything

````python
def log_event(event, *details, **meta):
    line = f"[{event}]"
    for detail in details:
        line += f" {detail}"
    for key, value in meta.items():
        line += f" {key}={value}"
    print(line)


log_event("LOGIN", "user=amina", success=True, attempts=1)
log_event("ERROR", "disk", "full")
````

## 8. Documenting functions

### Concept

A **docstring** is a string written first thing inside a function,
explaining what it does, its parameters and its return value. Tools (and
teammates) read it with `help()`. **Type hints** — optional annotations
like `name: str` — document what types are expected. They are *hints*:
Python does not enforce them, but they make intent explicit.

````python
def calculate_average(numbers):
    """Return the average of a list of numbers.

    Returns 0 for an empty list rather than crashing.
    """
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)


def shout(text: str) -> str:
    """Return text in capitals with an exclamation mark."""
    return text.upper() + "!"


print(calculate_average([2, 4, 6]))
print(shout("well done"))
````

````python
help(calculate_average)
````

## 9. Function composition

### Concept

Functions call other functions — small pieces combining into big
behaviour. Build small, single-purpose functions, then compose them.

````python
def calculate_total(prices):
    """Return the sum of all prices."""
    return sum(prices)


def calculate_average(prices):
    """Return the mean price."""
    return calculate_total(prices) / len(prices)      # reuse!


def price_report(prices):
    """Print a composed summary."""
    print(f"Items : {len(prices)}")
    print(f"Total : {calculate_total(prices):.2f}")
    print(f"Mean  : {calculate_average(prices):.2f}")


price_report([120, 80, 45])
````

Note how `calculate_average` doesn't recompute — it *calls*
`calculate_total`. Fix the total once, and every consumer is fixed.

## 10. Recursion — functions that call themselves

### Concept

A **recursive** function solves a problem by calling itself on a smaller
version of the problem. Every recursion needs:

1. a **base case** — the smallest problem, solved without recursing;
2. a **recursive case** — reduce the problem, call yourself again.

Without a base case the calls never stop — Python raises
`RecursionError` (the stack of waiting calls overflows).

### Example 1 — Countdown

````python
def countdown(n):
    if n == 0:                    # BASE CASE: stop here
        print("Lift off!")
        return
    print(n)
    countdown(n - 1)              # RECURSIVE CASE: smaller problem


countdown(3)
````

### How the calls actually work

Each call gets its *own* memory frame on the **call stack** and waits for
the next call to finish:

```text
countdown(3)  prints 3, waits for countdown(2)
  countdown(2)  prints 2, waits for countdown(1)
    countdown(1)  prints 1, waits for countdown(0)
      countdown(0)  prints Lift off! and RETURNS
    ... then each waiting call finishes, innermost first
```

### Example 2 — Factorial

````python
def factorial(n):
    """Return n! = n * (n-1) * ... * 1.  factorial(0) == 1."""
    if n == 0 or n == 1:          # base case
        return 1
    return n * factorial(n - 1)   # recursive case


print(factorial(5))     # 5*4*3*2*1
print(factorial(0))
````

### Example 3 — Fibonacci (and a warning)

````python
def fibonacci(n):
    """Return the n-th Fibonacci number (fib(1)=1, fib(2)=1)."""
    if n == 1 or n == 2:          # base cases
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)


for i in range(1, 8):
    print(fibonacci(i), end=" ")
print()
````

Recursive Fibonacci is *elegant but slow* — it recomputes the same values
over and over (Module 16 measures exactly how slow). A loop version is
faster; recursion is the clearer teaching example. **Do not overuse
recursion** — in this course, prefer loops unless the problem is naturally
recursive.

### Common Mistakes

```python
# 1. Missing base case:
# def boom(n): return boom(n - 1)      # RecursionError

# 2. Base case never reached:
# def bad(n):
#     if n == 0: return 1
#     return n * bad(n + 1)            # n grows away from 0!

# 3. Forgetting that each call has its OWN local variables.
```

## 11. Mini project — Calculator Using Functions

**Requirements**

1. Menu: add / subtract / multiply / divide / quit.
2. Each operation is its own function that **returns** (never prints) the result.
3. Division guards against zero *inside* its function by returning `None`
   (full exception handling arrives in Module 12).
4. The main loop reads numbers, calls the right function, prints the result.

**Pseudocode**

```text
FUNCTION add(a, b):        RETURN a + b
FUNCTION subtract(a, b):   RETURN a - b
FUNCTION multiply(a, b):   RETURN a * b
FUNCTION divide(a, b):     IF b == 0: RETURN None  ELSE RETURN a / b
LOOP:
    read choice; IF quit: break
    read a, b (converted with float())
    call the matching function, print result (or zero-division message)
```

````python
# ----- Calculator Using Functions -----
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        return None             # signal "no answer" to the caller
    return a / b


while True:
    print()
    print("--- Calculator ---")
    print("add / subtract / multiply / divide / quit")
    choice = input("Choice: ").strip().lower()

    if choice == "quit":
        print("Goodbye!")
        break
    elif choice in ("add", "subtract", "multiply", "divide"):
        first = float(input("First number: "))
        second = float(input("Second number: "))

        if choice == "add":
            result = add(first, second)
        elif choice == "subtract":
            result = subtract(first, second)
        elif choice == "multiply":
            result = multiply(first, second)
        else:
            result = divide(first, second)

        if result is None:
            print("Cannot divide by zero.")
        else:
            print(f"Result: {result}")
    else:
        print("Unknown choice.")
````

## Find the Error

**Error 1**

```python
double(5)

def double(x):
    return x * 2
```

**Error 2**

```python
def total(a, b=0, c):
    return a + b + c
```

**Error 3**
```python
def add(a, b):
    print(a + b)

result = add(2, 3) * 2
```

**Error 4**

```python
def countdown(n):
    print(n)
    countdown(n - 1)
```

**Error 5**

```python
def calc(x):
    return x + 1
    return x + 2

print(calc(1))
```

**Error 6**

```python
def greet():
    message = "hi"

print(message)
```

<details><summary>Answers</summary>

1. Called before defined — `NameError`. Define first, call after.
2. A parameter without a default follows one with a default — `SyntaxError`.
   Order: required parameters first.
3. `add` prints but returns `None`; `None * 2` is a `TypeError`. Return the
   sum instead of printing.
4. No base case — infinite recursion, ends in `RecursionError`. Stop when
   `n == 0`.
5. The second `return` is unreachable. Only `2` is printed (the code runs,
   but the second return is dead code — usually a logic bug).
6. `message` is local to `greet`; printing it outside raises `NameError`.

</details>

## 15. Summary

* `def` defines; the call runs. Parameters receive arguments.
* `return` gives a value back; `print` only shows. Functions without
  `return` give `None`.
* Positional, keyword, and default arguments control how calls bind to
  parameters; defaults come last.
* Return tuples for multiple values; unpack at the call site.
* Locals live and die inside calls; globals are readable everywhere but
  should be modified sparingly (if ever) via `global`.
* `*args` packs extra positionals into a tuple; `**kwargs` packs extra
  keywords into a dict.
* Docstrings (and optional type hints) document intent; `help()` reads them.
* Composition builds big behaviour from small functions; recursion needs a
  base case plus progress toward it.
