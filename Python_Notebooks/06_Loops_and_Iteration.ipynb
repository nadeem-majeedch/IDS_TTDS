## 1. Why loops?

### Concept

Suppose you must print "Happy Birthday!" ten times. Copy-pasting the same
line ten times works but is ugly, error-prone, and impossible for values
you only learn at run time ("print it 500 times", "print it once per
student"). **Loops** repeat a block of code automatically. One execution of
the block is called an **iteration**; running a loop is called
**iterating**; the repeated block is the **loop body**.

Manual repetition vs a loop:

````python
# Manual repetition — copy, paste, pray:
print("Happy Birthday!")
print("Happy Birthday!")
print("Happy Birthday!")

print("--- with a loop ---")

# A loop does the same in three lines, for ANY count:
for day_number in range(3):
    print("Happy Birthday!")
````

Change `range(3)` to `range(500)` and the loop happily does 500 iterations
— the manual version would need 500 pasted lines. That is why loops exist.

Python has two loop statements:

* **`for`** — repeat **once per item** of a sequence (a known count, a
  string's characters, ...). Use when you know *what* to iterate over.
* **`while`** — repeat **as long as** a condition stays True. Use when you
  know *when to stop*, but not how many rounds it will take.

## 2. The `for` loop and `range()`

### Syntax

```python
for variable in sequence:
    body                # indented 4 spaces, runs once per item
```

`range()` produces number sequences — three forms:

| Form | Meaning | Example |
|---|---|---|
| `range(stop)` | 0 up to stop−1 | `range(5)` → 0,1,2,3,4 |
| `range(start, stop)` | start up to stop−1 | `range(2, 6)` → 2,3,4,5 |
| `range(start, stop, step)` | every `step`-th | `range(0, 10, 2)` → 0,2,4,6,8 |

Like slices, the **stop value is excluded**.

### Example 1 — Beginner

````python
# Prints 0 1 2 3 4 — five iterations, stop value 5 is NOT printed.
for number in range(5):
    print(number)
````

### Explanation

Line by line:

* `for number in range(5):` — create the sequence 0,1,2,3,4. On each
  iteration, the next value is stored in `number` (the **loop variable**).
* `print(number)` — the body; runs once per value.
* The loop ends when the sequence is exhausted; execution continues below
  the loop.

### The three range forms in action

````python
print("range(5):")
for number in range(5):
    print(number, end=" ")
print()                              # move to a new line

print("range(2, 6):")
for number in range(2, 6):           # 2,3,4,5
    print(number, end=" ")
print()

print("range(0, 10, 2):")
for number in range(0, 10, 2):       # 0,2,4,6,8
    print(number, end=" ")
print()

print("range(5, 0, -1):")            # counting DOWN
for number in range(5, 0, -1):       # 5,4,3,2,1
    print(number, end=" ")
print()
````

**Note:** `range(5, 0, -1)` counts down but never reaches 0 — the stop is
excluded. Use `range(5, -1, -1)` to include it.

### Example 2 — Practical: countdown and timetable

````python
# A launch countdown:
for seconds_left in range(5, 0, -1):
    print(seconds_left, "...")
print("Liftoff!")
````

````python
# The multiplication table of 7:
table_number = 7

for multiplier in range(1, 11):
    result = table_number * multiplier
    print(f"{table_number} x {multiplier} = {result}")
````


## 3. Looping over strings

A string is a sequence of characters, so `for` can visit each character:

````python
word = "data"

for letter in word:
    print(letter)
````

````python
# With an index, use range(len(...)):
word = "data"

for index in range(len(word)):      # 0,1,2,3
    print(index, word[index])
````

````python
# Practical: count vowels (string + condition + loop — three modules united):
word = "programming"
vowel_count = 0

for letter in word:
    if letter in "aeiou":
        vowel_count += 1

print("Vowels in", word, ":", vowel_count)
````

## 4. The `while` loop

### Syntax

```python
while condition:
    body            # repeats while condition stays True
```

**The three ingredients of a well-behaved `while` loop:**

1. **Initialise** the control variable *before* the loop.
2. **Test** it in the condition.
3. **Update** it inside the body — or the loop never ends!

### Example 1 — Beginner: counting up

````python
count = 1                      # 1. initialise

while count <= 5:              # 2. test
    print("Count is", count)
    count += 1                 # 3. update — without this: infinite loop!
````

### Explanation

* Round 1: `count` is 1 → `1 <= 5` True → print, `count` becomes 2.
* ... continues until `count` is 6 → `6 <= 5` False → loop ends.
* The body ran exactly 5 times.

### Example 2 — Practical: saving until a goal is reached

````python
balance = 0
monthly_saving = 2500
target = 10000
months = 0

while balance < target:
    balance += monthly_saving
    months += 1

print(f"Target reached in {months} months with {balance} PKR")
````

A `while` is perfect here: we cannot know in advance how many months it
will take — the loop decides by itself.

### Infinite loops — and how to avoid them

````python
# DO NOT RUN the next three lines as-is: they never stop.
# while True:
#     print("forever")      # condition is always True, nothing changes it

# The fix: ensure SOMETHING in the body can make the condition False.
countdown = 3
while countdown > 0:
    print(countdown)
    countdown -= 1         # the escape hatch

print("Done")
````

If a notebook cell ever runs forever: **Kernel → Interrupt** (square stop
button) stops the runaway loop safely.

### Common Mistakes

```python
# 1. Forgetting the update:
# count = 1
# while count <= 5:
#     print(count)      # prints 1 forever

# 2. Updating in the wrong direction:
# count = 10
# while count > 0:
#     count += 1        # counts AWAY from 0 — never ends

# 3. Using == on floats in the condition:
# value = 0.0
# while value != 1.0:   # rounding may skip 1.0 exactly — use < or <=
```

## 5. `break`, `continue`, `pass`

### Concept

* **`break`** — leave the loop **immediately**; execution continues after
  the loop.
* **`continue`** — skip the rest of **this iteration**; jump straight to
  the next one.
* **`pass`** — do nothing; a placeholder where Python demands a statement.

### `break`

````python
# Stop as soon as we find the number 4:
for number in range(1, 11):
    if number == 4:
        print("Found 4 — stopping the search.")
        break
    print("Checked", number)
````

````python
# The classic use: searching.
secret = 7
guess = 1

while True:                    # "forever" — but with a way out:
    if guess * guess >= 49:    # crude square-root hunt
        print("Found a number whose square is >= 49:", guess)
        break
    guess += 1
````

### `continue`

````python
# Print only the odd numbers — even ones are skipped:
for number in range(1, 8):
    if number % 2 == 0:
        continue                 # skip the rest of THIS iteration
    print(number, "is odd")
````

### `pass`

`pass` is for *structure*, not behaviour — a stand-in where the syntax
requires a line but you have nothing to do yet:

````python
if 5 > 3:
    pass                        # plan: add real handling later
else:
    print("never printed")

print("pass caused no output at all")
````

### The three, side by side

````python
# break ends the loop entirely; continue skips one round; pass does nothing.
for number in range(1, 6):
    if number == 2:
        continue            # 2 is skipped
    if number == 4:
        break               # loop ends here; 4 and 5 never printed
    if number == 3:
        pass                # no effect at all
    print(number)
````

## 6. Counters and accumulators

These five patterns appear in virtually every program that processes data.
Learn them as *recipes*.

### 6.1 Counting — how many times does something happen?

````python
count = 0                       # the counter starts at 0

for number in range(1, 51):
    if number % 7 == 0:         # multiple of 7?
        count += 1              # one more found

print("Multiples of 7 up to 50:", count)
````

### 6.2 Summation — adding things up

````python
total = 0                        # the accumulator starts at 0

for price in (120, 85, 430, 60): # a tuple of prices to visit (a sequence)
    total += price

print("Total:", total)
````

### 6.3 Product — multiplying things together

Start at **1** (multiplying by 0 would erase everything):

````python
product = 1

for factor in (2, 3, 4):
    product *= factor

print("2 * 3 * 4 =", product)
````

### 6.4 Minimum and maximum

Carry the best-so-far; replace it whenever a better value appears:

````python
temperatures = (31, 28, 35, 33, 29)

highest = temperatures[0]        # start with the first value...
lowest = temperatures[0]

for temperature in temperatures:
    if temperature > highest:    # ...then challenge it each round
        highest = temperature
    if temperature < lowest:
        lowest = temperature

print("Highest:", highest)
print("Lowest:", lowest)
````

**Why start with the first element?** Starting `highest` at 0 fails when
all values are negative; starting it at 999999 fails for hot days. The
first element is always a safe starting champion.

### 6.5 Average — sum and count together

````python
marks = (78, 92, 65, 88)

marks_total = 0
marks_count = 0

for mark in marks:
    marks_total += mark
    marks_count += 1

average = marks_total / marks_count
print(f"Average of {marks_count} marks: {average:.2f}")
````

### 6.6 Frequency — counting categories

````python
sentence = "abracadabra"

count_a = 0
count_b = 0
count_other = 0

for character in sentence:
    if character == "a":
        count_a += 1
    elif character == "b":
        count_b += 1
    else:
        count_other += 1

print("a:", count_a, "| b:", count_b, "| other:", count_other)
````

(When you meet dictionaries in a later course, this pattern becomes a
one-liner — the logic you build here is exactly what it automates.)

### Example 3 — Challenge: factorial

The **factorial** of n (written n!) is 1 × 2 × 3 × ... × n. It combines
the product pattern with a range:

````python
n = 5
factorial = 1

for number in range(1, n + 1):   # 1,2,3,4,5
    factorial *= number

print(f"{n}! = {factorial}")
````

## 7. Nested loops

### Concept

A loop **inside** a loop. For every single step of the outer loop, the
inner loop runs **completely**. If the outer runs 3 times and the inner 4
times, the body executes 3 × 4 = 12 times.

### Example 1 — Beginner: pairs

````python
for row in range(1, 4):          # outer: 1,2,3
    for column in range(1, 4):   # inner: 1,2,3 — fully re-run per row
        print(f"({row},{column})", end=" ")
    print()                      # newline after each row
````

### Example 2 — Practical: all multiplication tables 1–3

````python
for table_number in range(1, 4):          # which table
    print(f"--- Table of {table_number} ---")
    for multiplier in range(1, 11):       # 1..10 within each table
        result = table_number * multiplier
        print(f"{table_number} x {multiplier} = {result}")
````

### Star and number patterns (a classic exercise, not the main dish)

````python
# Right triangle of stars: row 1 has 1 star, row 2 has 2, ...
size = 5

for row in range(1, size + 1):
    print("*" * row)             # string repetition from Module 4!
````

````python
# Number triangle:
for row in range(1, 6):
    for number in range(1, row + 1):
        print(number, end=" ")
    print()
````

````python
# Grid-style output: a 3x4 board of coordinates with row/col layout.
for row in range(1, 4):
    for column in range(1, 5):
        print(f"r{row}c{column}", end=" ")
    print()
````

With `end=" "` the coordinates of one row share a line; the bare `print()`
after the inner loop moves to the next row.

### Common Mistakes

```python
# 1. Using the inner variable outside its loop:
for row in range(3):
    for column in range(3):
        pass
    print(column)        # works but confusing: column is 2 here — do not

# 2. Wrong loop bound in patterns (range(1, size) misses the last row).

# 3. Forgetting that break leaves only the INNERMOST loop:
for row in range(3):
    for column in range(3):
        break            # exits the column loop only, every row!
```

## 8. Practical loop problems

### 8.1 Print even and odd numbers in a range

````python
print("Even numbers 1-10:")
for number in range(1, 11):
    if number % 2 == 0:
        print(number, end=" ")
print()

print("Odd numbers 1-10:")
for number in range(1, 11):
    if number % 2 != 0:
        print(number, end=" ")
print()
````

### 8.2 Reverse a number

Module 4 reversed text with slicing; numbers need arithmetic — the digit
trick from Module 2 inside a `while` loop:

````python
number = 4793
reversed_number = 0
remaining = number

while remaining > 0:
    last_digit = remaining % 10            # take the last digit...
    reversed_number = reversed_number * 10 + last_digit   # ...append it
    remaining //= 10                       # chop it off

print(number, "reversed is", reversed_number)
````

**Trace for 4793:** digit 3 → rev=3, remaining=479 → digit 9 → rev=39,
remaining=47 → digit 7 → rev=397, remaining=4 → digit 4 → rev=3974,
remaining=0 → stop.

### 8.3 Count digits

````python
number = 9025
digit_count = 0
remaining = number

while remaining > 0:
    digit_count += 1
    remaining //= 10

print(number, "has", digit_count, "digits")
````

### 8.4 Prime number check

A **prime** has exactly two divisors: 1 and itself. Strategy: try every
divisor from 2 up to number−1; if any divides evenly, it is *not* prime:

````python
number = 29
is_prime = True                    # innocent until proven guilty

for divisor in range(2, number):
    if number % divisor == 0:      # found another divisor!
        is_prime = False
        break                      # no need to keep looking

if number < 2:
    is_prime = False               # 0 and 1 are not prime

print(number, "is prime:", is_prime)
````

````python
# All primes below 30 — nested logic + accumulator + conditions:
print("Primes below 30:")
for candidate in range(2, 30):
    is_prime = True
    for divisor in range(2, candidate):
        if candidate % divisor == 0:
            is_prime = False
            break
    if is_prime:
        print(candidate, end=" ")
print()
````

### 8.5 Fibonacci sequence

Each Fibonacci number is the sum of the two before it: 0, 1, 1, 2, 3, 5,
8, ... The trick is updating **two** variables per round:

````python
terms = 8
previous = 0
current = 1

for term_number in range(terms):
    print(previous, end=" ")
    next_value = previous + current   # compute the next BEFORE overwriting
    previous = current                # shift the window forward
    current = next_value
print()
````

**Order matters:** if you wrote `previous = current` *before* computing
`next_value`, you would lose the old `previous`. Compute first, shift
second.

### 8.6 Find maximum and minimum with a loop

Section 6.4 showed the pattern on a tuple; here the values arrive one by
one inside the loop (like readings from a sensor):

````python
# Simulated readings — the loop must handle each as it comes:
readings = (21, 19, 24, 30, 17, 24)

highest_so_far = None      # nothing seen yet
lowest_so_far = None

for reading in readings:
    if highest_so_far is None or reading > highest_so_far:
        highest_so_far = reading
    if lowest_so_far is None or reading < lowest_so_far:
        lowest_so_far = reading

print("Max:", highest_so_far, "| Min:", lowest_so_far)
````

(This is also the first honest use of `is None` from Module 2!)

### 8.7 Guessing game (`while` + conditions + break)

````python
# The "computer" picked a secret; the loop plays the user's guesses.
secret_number = 42
guesses = (50, 20, 42)          # simulated guesses, in order

for guess in guesses:
    if guess == secret_number:
        print("Correct! The secret was", secret_number)
        break
    elif guess > secret_number:
        print(guess, "is too high")
    else:
        print(guess, "is too low")
````

### 8.8 Menu-driven program skeleton

A **sentinel value** (here `"4"`) tells the loop when to stop — a vital
real-world pattern:

````python
# Simulated menu choices (interactive input version in the mini project):
menu_choices = ("1", "2", "1", "4")

for choice in menu_choices:
    if choice == "1":
        print(">> Balance: 5000 PKR")
    elif choice == "2":
        print(">> Deposit recorded")
    elif choice == "3":
        print(">> Withdrawal recorded")
    elif choice == "4":
        print(">> Goodbye!")
        break
    else:
        print(">> Unknown option:", choice)
````

With real input you would write `while choice != "4":` and read inside the
loop — same skeleton, one line changed.

**Error 2**

```python

    print(count)
```

**Error 3**

<details><summary>Answers</summary>

2. `count` is never decreased — infinite loop; add `count -= 1` inside.
3. `total += number` lost its indentation, so it is outside the loop (and
