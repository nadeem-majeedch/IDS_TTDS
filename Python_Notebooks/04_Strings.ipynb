## 1. What is a string?

### Concept

A **string** is a sequence of characters: letters, digits, spaces, symbols.
Text of any kind — a name, a sentence, a whole paragraph — is stored as a
string. The type name is `str`.

### Syntax — creating strings

```python
single = 'single quotes work'
double = "double quotes too"
triple = """triple quotes can span
multiple lines"""
empty = ""                 # a string with no characters
```

### Example 1 — Beginner

````python
full_name = "Amina Khan"
single_line = 'Python is fun'
paragraph = """This string
spans two lines."""
no_characters = ""

print(full_name)
print(single_line)
print(paragraph)
print(len(no_characters))    # len() counts characters -> 0
````

**Strings are literals too.** A value written directly in code (like
`"Amina Khan"` or `42`) is called a **string literal** when it is text.
The quotes are not part of the string — they tell Python where it starts
and ends.

Why two kinds of quotes? So you can nest them:

````python
print("It's a beautiful day")            # apostrophe inside double quotes
print('She said "hello" to me')          # double quotes inside single quotes
print("She said \"hello\" politely")     # or escape them with backslash
````

### Escape sequences (recap from Module 3)

````python
print("Line 1\nLine 2")          # \n newline
print("Column A\tColumn B")      # \t tab
print("Backslash: \\")           # \\ prints one backslash
````

### Example 2 — Practical: building a label

````python
product = "University Notebook"
pages = 200

label = product + " - " + str(pages) + " pages"
print(label)
````

(f-strings, shown in section 8, make this much cleaner.)

## 2. Indexing

### Concept

Each character in a string has a position number called an **index**.
Python counts from **0**:

```text
 string:   P   y   t   h   o   n
 index:    0   1   2   3   4   5
 negative: -6  -5  -4  -3  -2  -1
```

* **Positive indexing** starts at 0 from the left.
* **Negative indexing** starts at -1 from the right — handy for "the last
  character" without knowing the length.

### Syntax

```python
string[index]
```

### Example 1 — Beginner

````python
word = "Python"

print(word[0])      # first character
print(word[1])      # second character
print(word[5])      # sixth (last) character
````

````python
print(word[-1])     # last character
print(word[-2])     # second to last
print(word[-6])     # first character, counted from the right
````

### Explanation

`word[0]` asks the string for the character at position 0 — the result is a
new, one-character string. `word[-1]` counts one from the end.

### Common Mistakes

```python
word = "Python"
# word[6]          # IndexError: string index out of range (valid: 0..5)
# word["1"]        # TypeError: indices must be integers
```

### Example 3 — Challenge: last character of any word

````python
# The last character without knowing the word in advance:
any_word = "Karachi"

last_character = any_word[len(any_word) - 1]   # len = 7, index 6
print(last_character)

print(any_word[-1])        # the idiomatic way — same result
````

## 3. Slicing

### Concept

A **slice** extracts a substring — a range of characters. This is one of
Python's most-loved features.

### Syntax

```python
string[start:stop]          # from index start up to (NOT including) stop
string[start:stop:step]     # same, taking every `step`-th character
```

**The stop index is always EXCLUDED.** `word[0:3]` gives characters at
indices 0, 1, 2 — three characters, none numbered 3.

![Index diagram](https://raw.githubusercontent.com/python-assignments/assets/main/index_diagram.png)

*(If the image does not load, rely on the text diagram: indices 0..5 above
each letter of `Python`, negative indices -1..-6 below.)*

### Example 1 — Beginner

````python
word = "Python"

print(word[0:3])     # indices 0,1,2 -> "Pyt"  (3 is excluded!)
print(word[1:4])     # indices 1,2,3 -> "yth"
print(word[3:])      # from index 3 to the end -> "hon"
print(word[:3])      # from the start up to (not incl.) 3 -> "Pyt"
print(word[:])       # the whole string -> "Python"
````

### Explanation

* Omitted `start` means "from the beginning".
* Omitted `stop` means "to the very end".
* `[:]` copies the whole string.

### Example 2 — Practical: first name, last name, extension

````python
full_name = "Amina Khan"
file_name = "report.pdf"

first_name = full_name[:5]        # "Amina"
last_name = full_name[6:]         # "Khan"
extension = file_name[-3:]        # last 3 characters: "pdf"

print("First:", first_name)
print("Last: ", last_name)
print("Type: ", extension)
````

### Negative indexes and step values

````python
word = "Python"

print(word[-4:-1])    # from -4 up to (not incl.) -1 -> "tho"
print(word[0:6:2])    # every 2nd char: indices 0,2,4 -> "Pto"
print(word[::3])      # every 3rd char: indices 0,3 -> "Ph"
print(word[::-1])     # step -1 walks backwards -> "nohtyP" (reversed!)
````

`word[::-1]` — the famous Python idiom for reversing a string. With a
negative step, a sensible reading of `word[::-1]` is "walk from the end to
the start".

### Example 3 — Challenge: middle character(s)

````python
# The middle character of an odd-length word:
word = "Karachi"      # 7 letters
middle_index = len(word) // 2
print(word[middle_index])            # 'a' — index 3

# For even length, grab the two middle characters:
even_word = "Lahore"         # 6 letters
centre = len(even_word) // 2            # 6 // 2 = 3
print(even_word[centre - 1:centre + 1]) # indices 2 and 3 -> "ho"
````

### Slicing never raises IndexError

Unlike indexing, a slice that runs past the end is silently clipped:
`"Python"[3:100]` gives `"hon"`. Handy — but it can hide typos.

## 4. String operations

### Concatenation `+`

Joins strings into a new string. Both sides must be strings:

````python
first_part = "Data"
second_part = "Science"

joined = first_part + " " + second_part
print(joined)
````

````python
# "+" does NOT convert: gluing a number needs str() or an f-string
age = 19
print("Age: " + str(age))
print(f"Age: {age}")
````

### Repetition `*`

A string times an integer repeats it:

````python
divider = "-" * 30
print(divider)
print("Report card")
print(divider)
````

### Membership `in` / `not in`

Checks whether a substring occurs anywhere in the string (True/False):

````python
sentence = "Python is a popular language"

print("popular" in sentence)
print("Popular" in sentence)      # case matters!
print("java" not in sentence)
````

### Comparison `==`, `<`, `>` ...

Strings compare by **dictionary/alphabetical order** (technically by
Unicode code points — so all uppercase letters sort before lowercase):

````python
print("apple" == "apple")
print("apple" < "banana")
print("Zebra" > "apple")     # True: 'Z' (90) < 'a' (97)!
print("abc" == "ABC")        # False: case-sensitive
````

Practical consequence: before comparing user input, normalise case
(`.lower()`) — see section 7.

## 5. String immutability

### Concept

Strings are **immutable**: they cannot be changed after creation. Every
"modification" actually builds a **new** string.

### Example — what fails and what works

````python
word = "cat"
# The following line WOULD fail — try uncommenting it:
#word[0] = "b"        # TypeError: 'str' object does not support item assignment

new_word = "b" + word[1:]     # build a NEW string instead
print(new_word)
````

````python
# Even method calls return new strings; the original is untouched:
shout = "hello"
upper_shout = shout.upper()

print(shout)        # still "hello"
print(upper_shout)  # the new string "HELLO"
````

**Why it matters:** always *capture the result*: `name = name.strip()` —
not just `name.strip()` (which does nothing visible!).

## 6. `len()` — counting characters

`len()` is a built-in function (not a method — no dot) that returns the
number of characters, spaces included:

````python
print(len("Python"))
print(len("Amina Khan"))      # the space counts!
print(len(""))                # empty string -> 0
````

## 7. The important string methods

### Concept

A **method** is a function attached to a value, called with a dot:
`value.method()`. Strings have dozens; these are the essentials. Remember:
each one **returns a new string**.

### 7.1 Changing case: `lower()`, `upper()`, `capitalize()`, `title()`

````python
text = "hello world"

print(text.lower())        # all lowercase
print(text.upper())        # ALL UPPERCASE
print(text.capitalize())   # First letter capital, rest lower
print(text.title())        # Every Word Capitalised
````

````python
# The classic use: comparing user input case-insensitively.
answer = "YES"
print(answer.lower() == "yes")
````

### 7.2 Cleaning whitespace: `strip()`, `lstrip()`, `rstrip()`

````python
messy = "   data science   "

print(messy.strip())     # both ends
print(messy.lstrip())    # left end only
print(messy.rstrip())    # right end only
````

(The leading/trailing spaces are hard to see in output — the f-string
`|...|` trick in section 8 makes them visible.)

````python
messy = "   data science   "
print(f"|{messy.strip()}|")
print(f"|{messy.lstrip()}|")
print(f"|{messy.rstrip()}|")
````

### 7.3 Replacing: `replace(old, new)`

````python
quote = "I like Java"
fixed = quote.replace("Java", "Python")
print(fixed)
print("month".replace("m", "M", 1))   # replace only first occurrence
````

### 7.4 Splitting and joining: `split()`, `join()`

The dynamic duo of text processing: `split()` cuts a string into a list of
pieces; `join()` glues a list of strings into one string. (Lists are a
later topic — here just follow the pattern.)

````python
record = "Amina,19,BS Data Science"
pieces = record.split(",")       # cut at every comma
print(pieces)
print(len(pieces))

sentence = "split on whitespace"
print(sentence.split())          # no argument: cut at spaces
````

````python
glued = "_".join(["2026", "09", "11"])
print(glued)

normalised = " ".join(["extra", "spaces", "are", "gone"])
print(normalised)
````

### 7.5 Searching: `find()`, `index()`, `count()`, `startswith()`, `endswith()`

````python
text = "programming is creative programming"

print(text.find("is"))            # index of first occurrence, or -1 if absent
print(text.find("xyz"))           # not found -> -1
print(text.index("is"))           # like find, but RAISES an error if absent
print(text.count("programming"))  # how many times?
````

````python
file_name = "report_final.pdf"

print(file_name.startswith("report"))   # True
print(file_name.endswith(".pdf"))       # True
print(file_name.endswith(".docx"))      # False
````

`find()` vs `index()`: both give the position of the first occurrence, but
`find()` returns `-1` when absent while `index()` crashes with a
`ValueError`. Prefer `find()` for simple searches, or check membership
first with `in`.

### Example 2 — Practical: cleaning a CSV-style record

````python
raw_record = "  42,  Data Science , 3.45  "
cleaned = raw_record.strip()
fields = cleaned.split(",")

roll_number = fields[0].strip()
department = fields[1].strip()
cgpa_text = fields[2].strip()

print(roll_number, "|", department, "|", cgpa_text)
print(f"CGPA as number: {float(cgpa_text):.2f}")
````

### Example 3 — Challenge: initials generator

````python
full_name = "Amina Raza Khan"
words = full_name.split()          # split on whitespace

initials = ""
for word in words:                 # loops are Module 6 — just follow along
    initials = initials + word[0].upper() + "."

print(initials)
````

### Common Mistakes

```python
# 1. Calling a method and ignoring the result (immutability!):
name = "  Amina "
name.upper()              # result thrown away — nothing changes
name = name.upper()       # correct: capture the new string

# 2. Case-sensitive comparisons:
"YES" == "yes"            # False — normalise with .lower() first

# 3. Off-by-one in slices:
"Python"[0:3]             # "Pyt" — remember, stop is EXCLUDED

# 4. len on a number:
# len(42)                  # TypeError — len() works on strings, not ints

# 5. Using index() on missing text:
# "abc".index("z")         # ValueError — use find() or `in` first
```

## 8. String formatting with f-strings

f-strings (Module 3) shine with text. Name, student info, sentences,
numbers — all in one readable template:

````python
student_name = "Ayesha"
roll_number = "CS-2104"
marks = 87.5
attendance_percent = 0.92

report_line = f"{student_name} ({roll_number}) scored {marks:.1f} marks."
summary_line = f"Attendance: {attendance_percent:.0%}"

print(report_line)
print(summary_line)
````

````python
# Aligned columns with width specs:
print(f"{'Name':<10}{'Marks':>8}")
print(f"{'-' * 18}")
print(f"{'Amina':<10}{87.5:>8.1f}")
print(f"{'Bilal':<10}{72.0:>8.1f}")
````

## 9. Text-processing examples

### 9.1 Count characters and words

````python
sentence = "Python makes text processing easy"

character_count = len(sentence)
word_count = len(sentence.split())

print("Characters:", character_count)
print("Words:", word_count)
````

### 9.2 Search for a word (case-insensitive)

````python
paragraph = "Data Science combines statistics and programming."
search_word = "SCIENCE"

found = search_word.lower() in paragraph.lower()
print(found)

position = paragraph.lower().find(search_word.lower())
print(position)
````

### 9.3 Replace words

````python
report = "The results were bad. The bad trend must stop."
improved = report.replace("bad", "promising")
print(improved)
````

### 9.4 Normalise text (clean + lowercase + single spaces)

````python
messy_input = "   Hello   WORLD   from   Python   "

# Step 1: trim the ends, step 2: lowercase.
normalised = messy_input.strip().lower()
# Step 3: split on whitespace and rejoin with single spaces.
normalised = " ".join(normalised.split())

print(f"|{normalised}|")
````

### 9.5 Start/end checks for file names

````python
file_name = "semester_results_2026.pdf"

is_pdf = file_name.endswith(".pdf")
is_secret = file_name.startswith("secret_")
starts_with_digit = file_name[0].isdigit()

print("PDF file:", is_pdf)
print("Secret file:", is_secret)
print("Starts with a digit:", starts_with_digit)
````

### 9.6 Reverse a string

````python
original = "notebook"
reversed_text = original[::-1]
print(reversed_text)
````

### 9.7 Palindrome check

A **palindrome** reads the same forwards and backwards (`madam`, `noon`).
Real checks ignore case and spaces — prepare the text first, then compare
with its reverse:

````python
candidate = "Never odd or even"

prepared = candidate.lower().replace(" ", "")
is_palindrome = prepared == prepared[::-1]

print(f"'{candidate}' is a palindrome: {is_palindrome}")
````

````python
# A non-palindrome for contrast:
check_word = "python"
print(check_word == check_word[::-1])
````

### Example 3 — Challenge: word statistics

````python
report_text = "Data science is data driven and data minded"

words = report_text.split()
total_words = len(words)
data_mentions = report_text.lower().count("data")

print("Total words:", total_words)
print("'data' appears:", data_mentions, "times")
print("Longest-ish first word:", words[0])
````
