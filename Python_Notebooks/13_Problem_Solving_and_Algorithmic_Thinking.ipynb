## 1. The problem-solving methodology

### Concept

A repeatable process beats inspiration. Eleven steps, in order:

1. **Understand** the problem — restate it in your own words.
2. **Identify inputs** — what data arrives, and in what form?
3. **Identify outputs** — what exactly must come out?
4. **Identify constraints** — ranges, formats, special cases.
5. **Decompose** — break the problem into smaller sub-problems.
6. **Design the algorithm** — the precise sequence of steps.
7. **Write pseudocode** — the algorithm in structured English.
8. **Implement** — translate to Python, one sub-problem at a time.
9. **Test** — normal, boundary, and invalid cases.
10. **Debug** — trace, bisect, fix (section 6).
11. **Improve** — readability, names, structure (Module 16 goes deeper).

Beginners skip 1–7 and debug *design* errors at the keyboard. The steps
exist precisely to move thinking *before* typing, where it is cheap.

## 2. Worked example 1 — from problem to tested program

**Problem.** A class wants its marks analysed: read a list of marks
(−1 ends entry), print the highest, lowest, average, and how many passed
(≥ 50).

### Steps 1–4: understand, inputs, outputs, constraints

* **Restate:** repeatedly receive marks; when −1 appears, stop and report
  statistics.
* **Inputs:** integers 0–100, then the sentinel −1.
* **Outputs:** highest, lowest, average (1 decimal), pass count.
* **Constraints:** −1 is *data*, not a mark; the list may be empty (−1
  first!); marks outside 0–100 are ignored with a warning.

### Steps 5–7: decompose, algorithm, pseudocode

```text
marks <- empty list
LOOP:
    read text
    IF text == "-1": stop loop
    n <- int(text)                       # may fail
    IF 0 <= n <= 100: append n  ELSE: warn
IF marks is empty: print "no data"; STOP
highest <- max(marks); lowest <- min(marks)
average <- sum(marks) / len(marks)
passed <- count of marks >= 50
print all four
```

### Step 8: implement

````python
# Simulated typing session (a real program would use input()):
typed_marks = ["85", "42", "90", "77", "-1"]

marks = []
for text in typed_marks:
    if text == "-1":
        break
    try:
        value = int(text)
    except ValueError:
        print("Not a number:", text)
        continue
    if 0 <= value <= 100:
        marks.append(value)
    else:
        print("Ignored (0-100 only):", value)

if len(marks) == 0:
    print("No data entered.")
else:
    highest = max(marks)
    lowest = min(marks)
    average = sum(marks) / len(marks)
    passed = 0
    for mark in marks:
        if mark >= 50:
            passed += 1
    print(f"Highest: {highest}")
    print(f"Lowest : {lowest}")
    print(f"Average: {average:.1f}")
    print(f"Passed : {passed} of {len(marks)}")
````

### Steps 9–11: test, debug, improve

* **Normal:** the run above.
* **Boundary:** −1 immediately (empty case), 0 and 100 as marks, exactly 50
  (passes!).
* **Invalid:** "abc" → warning, ignored; 150 → ignored.
* **Improve:** the pass-count loop could use `sum(1 for ...)` — but plain
  loops are clearer at this stage (Module 16 revisits).

The empty case is the one beginners forget — we handled it *before* it
could crash `max([])`.

## 3. Pseudocode conventions

### Concept

Pseudocode is the algorithm written in structured English — no syntax
rules, but *sequential*, *indented*, and using the same three control
structures you know:

```text
SET total TO 0                     # assignment
FOR each item IN list:             # loop
    IF item > best THEN            # decision
        SET best TO item
OUTPUT best                        # result
```

Keep it language-neutral (READ/SET/FOR/IF/WHILE/OUTPUT) and one-line-per-
action. If your pseudocode needs Python tricks, it's Python — rewrite the
*idea*.

### Example — Practical: pseudocode to code, both shown

````python
# Pseudocode:
#   SET count TO 0
#   FOR each character IN text:
#       IF character is a digit THEN count <- count + 1
#   OUTPUT count

text = "Room 101, floor 3"
count = 0
for character in text:
    if character.isdigit():
        count += 1
print(count)
````

## 4. Flow representation (text-based flowcharts)

### Concept

A flowchart shows control flow: **ovals** start/end, **rectangles**
actions, **diamonds** decisions, **arrows** the path. In a notebook, a
text-flow diagram renders reliably everywhere (fancy diagram libraries
may not):

```text
        (START)
           |
    [ read password ]
           |
       < password == "open" ? >
       /                     \
     yes                      no
      |                        |
 [ print Welcome ]     [ print Denied ]
       \                     /
        ( END )
```

Rules of thumb: one decision per diamond; every arrow labelled on
branches; the flow *always* reaches an END.

### Example — Practical: the guess-limit loop as text flow

````python
# ```text
# (START)
#    |
# [ secret <- 7 ; tries <- 0 ]
#    |
#    +------< tries < 3 ? >------no----> (END: out of tries)
#    |            |
#    |           yes
#    |            |
#    |    [ guess <- next guess ]
#    |            |
#    |    < guess == secret ? >---yes---> [ print "Got it" ] ---> (END)
#    |            |
#    |           no
#    |            |
#    |    [ tries <- tries + 1 ]
#    |            |
#    +------------+
# ```
````

````python
secret = 7
guesses = [3, 9, 7]        # simulated player
tries = 0
outcome = "out of tries"

for guess in guesses:
    tries += 1
    if guess == secret:
        outcome = f"Got it in {tries} tries"
        break

print(outcome)
````

## 5. Tracing and dry runs

### Concept

A **dry run** executes the program *on paper*: a table with one column per
variable and one row per step. It finds logic errors without a computer
and is the fastest way to *understand* a loop you didn't write.

### Example — Beginner: dry run of a running total

````python
# total <- 0
# FOR number IN [3, 1, 4]:
#     IF number > total THEN total <- number
# OUTPUT total

# | step | number | total |
# |-----:|-------:|------:|
# |  1   |   3    |   3   |
# |  2   |   1    |   3   |
# |  3   |   4    |   4   |
#
# Prediction: 4. Now run:

total = 0
for number in [3, 1, 4]:
    if number > total:
        total = number
print(total)
````

### Example — Practical: a dry-run table with output column

````python
# for i in range(1, 4):
#     print(i, i * i)
#
# | step |  i | output |
# |-----:|---:|--------|
# |  1   |  1 | 1 1    |
# |  2   |  2 | 2 4    |
# |  3   |  3 | 3 9    |
#
# Prediction: three lines. Verify:

for i in range(1, 4):
    print(i, i * i)
````

**How to use dry runs for debugging:** when a program misbehaves, make the
table for a *small* input; walk it honestly (no assuming!); the first row
where paper and code disagree is your bug.

## 6. Debugging techniques

### Concept

Debugging is systematic search, not guesswork:

1. **Read the traceback** — last line names the exception; the `line N`
  pointers localise it.
2. **Reproduce small** — shrink the input until the bug is 3 lines.
3. **print() bisection** — print variables halfway; halve the search
  space each time.
4. **Check the assumptions** — is the list really non-empty? Is the type
   really int? Print `type(x)`, `len(x)`.
5. **Rubber-duck it** — explain line by line aloud; the bug confesses.

### Example — Practical: bisecting a wrong average

````python
# BUG REPORT: "average of [70, 80, 90] prints 240"
marks = [70, 80, 90]
total = 0
for mark in marks:
    total += mark
    print("debug: after", mark, "total =", total)     # bisection print
average = total * len(marks)        # suspicious line
print("average:", average)
````

The prints prove `total` reaches 240 correctly — so the bug is *after*
the loop. Reading `average = total * len(marks)` with fresh eyes: should
be `/`. This is bisection in action: the prints cut the program in half
and the bug could only live in one half.

## 7. Problem families — worked solutions

### 7.1 Number problems

#### Digit sum

````python
# Pseudocode: WHILE n > 0: total <- total + n % 10 ; n <- n // 10
number = 4923
total = 0
working = number
while working > 0:
    total += working % 10          # last digit
    working = working // 10        # drop last digit
print(total)
````

````python
# Dry run for 4923:  (3) 492 -> (2) 49 -> (9) 4 -> (4) 0 stop
#                    3 + 2 + 9 + 4 = 18  ✓
````

#### Reverse a number

````python
number = 1234
reversed_number = 0
working = number
while working > 0:
    reversed_number = reversed_number * 10 + working % 10
    working = working // 10
print(reversed_number)
````

#### Prime check — trial division to √n

````python
# Pseudocode:
#   IF n < 2 -> not prime
#   FOR d FROM 2 WHILE d * d <= n:
#       IF n % d == 0 -> not prime
#   otherwise prime

def is_prime(n):
    if n < 2:
        return False
    divisor = 2
    while divisor * divisor <= n:
        if n % divisor == 0:
            return False
        divisor += 1
    return True


candidates = [1, 2, 9, 17, 25, 97]
results = []
for candidate in candidates:
    results.append((candidate, is_prime(candidate)))
print(results)
````

#### Armstrong-style (3-digit): number == sum of cubes of its digits

````python
number = 153
digit_total = 0
working = number
while working > 0:
    digit = working % 10
    digit_total += digit ** 3
    working //= 10
print(number == digit_total)
````

````python
# 1³ + 5³ + 3³ = 1 + 125 + 27 = 153 ✓ — try 154 in a scratch cell (False).
````

### 7.4 List problems

#### Second largest without sorting

````python
marks = [12, 45, 2, 45, 41, 7]

largest = None
second = None
for mark in marks:
    if largest is None or mark > largest:
        second = largest            # old champion demoted
        largest = mark
    elif mark != largest and (second is None or mark > second):
        second = mark

print(largest, second)
````

(The `mark != largest` guard handles the duplicate 45s — exactly the trap

### 7.5 Searching problems — linear vs binary

#### Linear search

````python
def linear_search(items, target):
    for index in range(len(items)):
        if items[index] == target:
            return index
    return -1


print(linear_search([4, 8, 15, 16, 23], 15))
print(linear_search([4, 8, 15, 16, 23], 99))
````

#### Binary search — only on *sorted* data (see Module 16 for why it wins)

````python
def binary_search(sorted_items, target):
    low = 0
    high = len(sorted_items) - 1
    while low <= high:
        middle = (low + high) // 2
        if sorted_items[middle] == target:
            return middle
        elif sorted_items[middle] < target:
            low = middle + 1        # discard the left half
        else:
            high = middle - 1       # discard the right half
    return -1


data = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]
print(binary_search(data, 23))
print(binary_search(data, 24))
````

### 7.8 Menu-driven programs

The Module 6/11 pattern, now with functions and validation — the skeleton
of every lab project:

````python
# Pseudocode:
#   LOOP:
#     show menu; read choice
#     IF choice in menu: dispatch; ELSE message
#     IF choice == quit: break

def process(numbers, action):
    if action == "sum":
        return sum(numbers)
    if action == "max":
        return max(numbers) if len(numbers) > 0 else None
    return None


numbers = [4, 8, 15]
print(process(numbers, "sum"), process(numbers, "max"))
````

````python
# Guided 1 — the methodology in miniature: mean without the mean().
# P: inputs [marks], output mean; plan: total, count, divide.
marks = [65, 70, 85]
total = 0
count = 0
for mark in marks:
    total += mark
    count += 1
print(total / count)
````

````python
# Guided 2 — plan-then-code: how many days until the list is used up?
# P: one biscuit pack feeds 4 students/day; 3 packs; 90 students.
#    Inputs: packs=3, per_pack=4, students=90. Output: days of supply.
packs = 3
biscuits = packs * 4
students = 90
days = 0
while biscuits >= students:
    biscuits -= students
    days += 1
print(days, "full day(s);", biscuits, "left over")
````

Look at that output: **0 days, 12 "left over"**. Technically the loop told
the truth — 12 biscuits cannot feed 90 students even once — but a program
that reports a happy-looking "12 left over" for a completely impossible
supply is *misleading by design*. The methodology catches it: at step 4
(constraints) we should have asked "what if supply < one day's need?" A
dry run of the guard with biscuits=12, students=90 shows the loop body
never executes. Fixing properly:

````python
packs = 3
biscuits = packs * 4
students = 90
if biscuits >= students:
    days = 0
    while biscuits >= students:
        biscuits -= students
        days += 1
    print(days, "full day(s);", biscuits, "left over")
else:
    print("Not enough supply: need", students - biscuits, "more biscuits.")
````

<details><summary>Answers</summary>

1. Off-by-one: `range(1, 5)` stops at 4. Use `range(1, 6)`.
2. `total = 0` *inside* the loop resets every pass — returns one mark.
   Move the reset above the loop.
3. Mutating while iterating skips the shifted 3 → prints `[1, 2, 3]`
   minus one 2 as `[1, 3]`? Trace it: it actually prints `[1, 3]` only by
   luck of positions; the *robust* fix is building a new list.
4. Division produces a float quotient; evenness is `n % 2 == 0`.
5. The last valid index is `len(text) - 1`.
6. `max([])` raises ValueError — guard with `if values:` first.

</details>
