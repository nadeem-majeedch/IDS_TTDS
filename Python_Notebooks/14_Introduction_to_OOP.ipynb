## 1. Procedural vs object-oriented thinking

### Concept

**Procedural** (Modules 1–13): the unit of organisation is the *function*;
data is passed around.

```python
student = {"name": "Ali", "marks": 78}      # data
def promote(record): ...                    # behaviour lives apart
```

**Object-oriented:** the unit is the *object* — data + behaviour glued:

```python
student = Student("Ali", 78)     # knows its own data...
student.promote()                # ...and its own behaviour
```

### Why objects?

When a program manages many "things of the same kind" (students, accounts,
products), classes keep each thing's data and the operations on it in one
place. You get:

* a **blueprint** (the class) written and tested once;
* **instances** (objects) that each carry their own data;
* operations that cannot forget which data they belong to (no more
  passing the right dict to the wrong function).

## 2. Classes and objects — the analogy

A **class** is a cookie cutter; **objects** are the cookies. The cutter
defines shape (attributes) and decoration rules (methods); every cookie is
independent — bite one, the others don't notice.

```text
class Student  ------------------  the blueprint
  attributes: name, marks        "every student HAS a name and marks"
  methods:    introduce, grade   "every student CAN introduce and grade"

student1 = Student("Ali", 78)    two separate cookies,
student2 = Student("Sara", 91)   same cutter
```

## 3. Defining a class — `__init__` and `self`

### Concept

* `class Student:` starts the blueprint.
* `__init__` is the **constructor**: Python calls it automatically on
  every `Student(...)` creation, to set up the new object's attributes.
* `self` is the object being built/used — the first parameter of every
  method, filled in *automatically* by Python.

### Syntax

```python
class ClassName:
    def __init__(self, param1, param2):
        self.attribute1 = param1
        self.attribute2 = param2
```

### Example 1 — Beginner

````python
class Student:
    def __init__(self, name, marks):
        self.name = name        # this object's name
        self.marks = marks      # this object's marks


student1 = Student("Ali", 78)       # __init__ runs; self = the new object
student2 = Student("Sara", 91)

print(student1.name, student1.marks)
print(student2.name, student2.marks)
````

### Explanation, line by line

* `class Student:` — define the blueprint; the body is indented.
* `def __init__(self, name, marks):` — constructor: Python passes `self`
  automatically; you supply `name` and `marks` at creation.
* `self.name = name` — "store `name` *on this object*". The left side is
  the object's attribute; the right side the parameter received.
* `Student("Ali", 78)` — create: allocate the object, run `__init__` with
  `self` = that object, return it to `student1`.

`student1` and `student2` are independent — different attributes, same
blueprint.

### Example 2 — Practical: attributes can be computed at birth

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
        self.area = width * height          # computed once, at construction


small = Rectangle(3, 4)
large = Rectangle(10, 2)
print(small.area, large.area)
````

## 4. Methods — behaviour with access to data

### Concept

A **method** is a function defined inside a class. It takes `self` first,
so it can read and update the object's own attributes.

### Example 1 — Beginner: introduce and grade

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def introduce(self):
        return f"Hi, I am {self.name}."

    def letter_grade(self):
        if self.marks >= 80:
            return "A"
        elif self.marks >= 60:
            return "B"
        return "C"


ali = Student("Ali", 78)
sara = Student("Sara", 91)

print(ali.introduce())
print(ali.letter_grade(), sara.letter_grade())
````

### Explanation

* `ali.introduce()` means: call `introduce` with `self` = `ali` — the
  method reads *ali's* `self.name`.
* The two calls of `letter_grade` run the same code on different data —
  that is the whole point.

### Example 2 — Practical: methods that update state

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
            return self.balance
        self.balance -= amount
        return self.balance


account = BankAccount("Amina", 100)
print(account.deposit(50))
print(account.withdraw(30))
print(account.withdraw(500))
````

(`withdraw` refuses politely — a hand-made guard. Module 12's exceptions
would be the grown-up version: `raise ValueError(...)`.)

### Common Mistakes

```python
# 1. Forgetting self in the parameter list:
#    def introduce(): return f"Hi {name}"   # TypeError: missing self

# 2. Forgetting self. on attributes inside methods:
#    return name           # NameError — the attribute is self.name

# 3. Passing self yourself:
#    ali.introduce(ali)    # TypeError — Python already passed it
```

## 5. Many objects — objects in collections

Objects are values like any other: store them in lists, loop over them,
call their methods (Module 7 + 14 combined):

````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def passed(self):
        return self.marks >= 50


students = [
    Student("Ali", 78),
    Student("Sara", 91),
    Student("Omar", 42),
]

for student in students:
    status = "PASS" if student.passed() else "FAIL"
    print(f"{student.name:<6} {student.marks:>3}  {status}")
````

````python
average = 0
total = 0
for student in students:
    total += student.marks
average = total / len(students)
print(f"Class average: {average:.1f}")
````

## 6. Encapsulation — the concept and the convention

### Concept

**Encapsulation** means an object guards its own data: outside code uses
methods (`account.deposit(50)`) instead of poking attributes directly.
Benefits: the class can validate changes (negative deposit?) and can
change its internals later without breaking users.

Python's style is *honest, not forceful*: attributes named with a leading
underscore (`self._hidden`) are a **convention** meaning "internal —
hands off"; the language does not enforce it. A double underscore
(`__secret`) triggers name-mangling — mention only, Module 16's reading
list covers details.

````python
class Timer:
    def __init__(self):
        self._elapsed = 0            # _internal: do not touch from outside

    def tick(self, seconds=1):
        if seconds < 0:
            print("Ignored negative tick.")
            return
        self._elapsed += seconds

    def elapsed(self):
        return self._elapsed


timer = Timer()
timer.tick(30)
timer.tick(10)
print(timer.elapsed())
````

## 7. Inheritance — Person → Student

### Concept

A **child class** inherits the attributes and methods of a **parent
class**, then adds or overrides. `class Student(Person)` reads "a Student
*is a* Person" — the classic test for inheritance.

### Example

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"I am {self.name}, {self.age} years old."


class Student(Person):                    # Student IS-A Person
    def __init__(self, name, age, program):
        super().__init__(name, age)       # let Person set name and age
        self.program = program            # Student adds its own

    def introduce(self):                  # OVERRIDE with more detail
        base = super().introduce()
        return base + f" I study {self.program}."


stranger = Person("Bilal", 40)
freshman = Student("Alia", 19, "Data Science")

print(stranger.introduce())
print(freshman.introduce())
````

### Explanation

* `super().__init__(...)` — call the parent's constructor so name/age are
  set exactly as before; then the child adds `program`.
* **Overriding**: `Student.introduce` replaces `Person.introduce`; the
  child still *reuses* the parent's wording via `super()`.
* The child inherits everything else (future `Person` methods would be
  callable on students too).

## 8. Polymorphism — same call, fitting behaviour

### Concept

**Poly-morphism** = "many shapes": different objects respond to the *same
method name* each in their own way. Code that calls `introduce()`
therefore works on any mix of People and Students — no type checks.

````python
class Person:
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I am {self.name}."


class Student(Person):
    def __init__(self, name, program):
        super().__init__(name)
        self.program = program

    def introduce(self):
        return f"I am {self.name}, studying {self.program}."


class Teacher(Person):
    def __init__(self, name, subject):
        super().__init__(name)
        self.subject = subject

    def introduce(self):
        return f"I am {self.name}, teaching {self.subject}."


people = [
    Person("Guest"),
    Student("Alia", "Data Science"),
    Teacher("Dr Khan", "Statistics"),
]

for person in people:
    print(person.introduce())
````

One loop, three behaviours — that is polymorphism. (You have already used
it unknowingly: `len()` works on strings *and* lists; both "know how" to
answer.)

## 9. When to use classes (and when not)

**Use a class when:**

* several pieces of data always travel together *and* have operations
  (`BankAccount`: owner, balance, deposit, withdraw);
* you need many instances with independent state;
* the domain naturally speaks of "things" (Student, Product, Timer).

**Skip classes when:**

* a single function does the job (`celsius_to_fahrenheit`);
* data is one fixed record — a tuple or dict is clearer;
* you would create exactly one object with one method — that is a
  function wearing a costume.

Rule of thumb: **functions organise behaviour; classes organise data plus
its behaviour.**

## 10. Mini project — Student Management Using Classes

**Requirements**

1. `Student` class: name, marks (list), methods `add_mark`,
   `average`, `result` ("PASS"/"FAIL" via ≥ 50 average).
2. `StudentManager` class: keeps students, `add(student)`, `search(name)`,
   `report()` printing every student + class average.
3. Validation: marks 0–100 via `raise` (Module 12 in the OOP world).

**Pseudocode**

```text
CLASS Student:  init(name); add_mark(m) validates & appends;
                average(); result()
CLASS StudentManager: students <- []
    add(s); search(name) -> student or None
    report(): FOR each student print name, average, result;
              print class average
```

````python
class Student:
    def __init__(self, name):
        self.name = name
        self._marks = []                  # encapsulated list

    def add_mark(self, mark):
        if mark < 0 or mark > 100:
            raise ValueError(f"marks must be 0-100, got {mark}")
        self._marks.append(mark)

    def average(self):
        if len(self._marks) == 0:
            return None
        return sum(self._marks) / len(self._marks)

    def result(self):
        average = self.average()
        if average is None:
            return "NO MARKS"
        return "PASS" if average >= 50 else "FAIL"


class StudentManager:
    def __init__(self):
        self.students = []

    def add(self, student):
        self.students.append(student)

    def search(self, name):
        for student in self.students:
            if student.name.lower() == name.lower():
                return student
        return None

    def report(self):
        print(f"{'Name':<8}{'Avg':>7}  Result")
        for student in self.students:
            average = student.average()
            shown = f"{average:.1f}" if average is not None else "  -"
            print(f"{student.name:<8}{shown:>7}  {student.result()}")
        averages = []
        for student in self.students:
            if student.average() is not None:
                averages.append(student.average())
        if len(averages) > 0:
            print(f"Class average: {sum(averages) / len(averages):.1f}")


manager = StudentManager()
manager.add(Student("Ali"))
manager.add(Student("Sara"))
manager.add(Student("Omar"))

manager.search("Ali").add_mark(78)
manager.search("Ali").add_mark(90)
manager.search("Sara").add_mark(91)
try:
    manager.search("Omar").add_mark(150)
except ValueError as error:
    print("Rejected:", error)
manager.search("Omar").add_mark(45)

manager.report()
````

### Guided exercises

````python
# Guided 1 — a Counter class (state + behaviour in 8 lines).
class Counter:
    def __init__(self):
        self.value = 0

    def increment(self, step=1):
        self.value += step
        return self.value


counter_a = Counter()
counter_b = Counter()
counter_a.increment()
counter_a.increment(5)
counter_b.increment()
print(counter_a.value, counter_b.value)      # independent state!
````

````python
# Guided 2 — inheritance with override, the minimal example.
class Vehicle:
    def __init__(self, wheels):
        self.wheels = wheels

    def describe(self):
        return f"{self.wheels} wheels"


class Bicycle(Vehicle):
    def __init__(self):
        super().__init__(2)

    def describe(self):
        return "A bicycle: " + super().describe()


print(Bicycle().describe())
````

````python
# Guided 3 — objects in a list, polymorphic loop.
class Shape:
    def area(self):
        return 0


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2


shapes = [Square(4), Circle(1)]
for shape in shapes:
    print(round(shape.area(), 2))
````

## Find the Error

**Error 1**

```python
class Cat:
    def __init__(name):
        self.name = name
```

**Error 2**

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def speak():
        return "woof"

print(Dog("Rex").speak())
```

**Error 3**

```python
class Counter:
    def __init__(self):
        self.count = 0

    def bump():
        self.count += 1

c = Counter()
c.bump()
```

**Error 4**

```python
class A:
    def __init__(self, x):
        self.x = x

a = A(5)
print(a.x(2))       # meant: a.x * 2
```

**Error 5**

```python
class Parent:
    def __init__(self, name):
        self.name = name

class Child(Parent):
    def __init__(self, name, age):
        self.age = age        # forgot super().__init__(name)
```

<details><summary>Answers</summary>

1. Every method needs `self` first: `def __init__(self, name):`.
2. `speak` needs `self`: `def speak(self):` — else TypeError on call.
3. Same bug in `bump` — missing `self`.
4. `x` is an int attribute, not callable: `print(a.x * 2)`.
5. The child never sets `name`; `child.name` raises AttributeError. Add
   `super().__init__(name)` first.

</details>

## 14. Summary

* Classes bundle **data (attributes)** with **behaviour (methods)**;
  objects are independent instances.
* `__init__` initialises each new object; `self` is the object itself,
  passed automatically.
* Methods keep behaviour with the data it belongs to — no more passing
  the wrong dict to the right function.
* Encapsulation guards state behind methods; `_underscore` is the
  "internal" convention.
* Inheritance (`Child(Parent)` + `super()`) shares code down an is-a
  chain; overriding specialises; polymorphism lets one loop call many
  behaviours.
* Use classes for stateful things; use functions for stateless
  calculations. Choose deliberately.
