# PART A — Tuples

## A1. What is a tuple?

### Concept

A **tuple** is an ordered collection of values that, once created, **cannot
change**. It is a "locked list". Where a list is a whiteboard, a tuple is a
laminated card: readable forever, writable never.

### Why do we need it?

Some data is fixed by its nature: a date of birth, a coordinate (latitude,
longitude), a student's (name, age, programme). Locking such data into a
tuple tells every reader — and every future you — that this value will not
be mutated somewhere else in the program.

### Syntax

```python
point = (3, 5)                 # parentheses, commas
single = (42,)                 # ONE element needs the trailing comma!
```

````python
point = (3, 5)
single = (42,)
not_a_tuple = (42)              # just parentheses around a number!

print(point)
print(single)
print(not_a_tuple, type(not_a_tuple))

coordinates_without_parens = 10, 20     # even the parentheses are optional
print(coordinates_without_parens)
````

### Explanation

* The **comma** makes the tuple, not the parentheses — `(42)` is just 42 in
  brackets, while `(42,)` is a one-element tuple.
* Python prints tuples with parentheses so you can recognise them.

### When is a tuple preferable to a list?

* The data is a **fixed record**: RGB colours `(255, 0, 0)`, weekdays,
  (name, marks) pairs.
* You want to make sure **no code can accidentally modify** the values.
* Later (Module 16): tuples can be used where lists cannot, e.g. as
  dictionary keys.

## A2. Indexing and slicing

Tuples behave exactly like lists for *reading* — same indexes, same slices.

````python
student = ("Ali", 21, "Data Science")

print(student[0])       # first element
print(student[-1])      # last element
print(student[0:2])     # slice -> a NEW smaller tuple
print(len(student))     # number of elements
````

````python
for field in student:               # iteration works the same as lists
    print("field:", field)
````

## A3. Immutability

### Concept

**Immutable** means "cannot be changed after creation". Strings and tuples
share this property; lists, sets and dictionaries do not.

````python
weekdays = ("Mon", "Tue", "Wed", "Thu", "Fri")

# The next line would raise an error — read it, then comment it out to test:
#weekdays[0] = "Monday"       # TypeError: 'tuple' object does not support item assignment

print(weekdays)
````

Want a "changed" tuple? Build a new one:

````python
weekdays = ("Mon", "Tue", "Wed", "Thu", "Fri")
longer = ("Monday",) + weekdays[1:]      # new tuple from pieces
print(longer)
````

## A4. Tuple methods — `count()` and `index()`

A tuple has only these two methods (no methods that change it, of course):

````python
rolls = (1, 6, 3, 6, 6, 2)

print(rolls.count(6))     # how many sixes?
print(rolls.index(3))     # position of the first 3
print(6 in rolls)         # membership, like lists
print(4 not in rolls)
````

## A5. Tuple unpacking

### Concept

**Unpacking** assigns each element of a tuple to its own variable in one
line. The number of variables must match the number of elements.

### Syntax

```python
name, age, program = student
```

````python
student = ("Ali", 21, "Data Science")

name, age, program = student

print(name)
print(age)
print(program)
````

Unpacking also makes swapping variables trivial — and works in loops:

````python
red = 255
green = 0
red, green = green, red          # swap without a temporary variable
print(red, green)
````

````python
pairs = [("Ali", 78), ("Sara", 91), ("Omar", 84)]
for name, marks in pairs:        # unpack directly in the for loop
    print(f"{name}: {marks}")
````

## A6. Converting between lists and tuples

````python
marks_list = [80, 90, 85]
marks_tuple = tuple(marks_list)      # list -> tuple (a frozen snapshot)

back_to_list = list(marks_tuple)     # tuple -> list (editable again)

marks_list.append(95)                # the list changes...
print("list :", marks_list)
print("tuple:", marks_tuple)         # ...the tuple does not care
print("back :", back_to_list)
````

```python

# 2. Trying to modify:
# t[0] = "new"     # TypeError

# 3. Wrong number of variables when unpacking:
# name, age = ("Ali", 21, "DS")    # ValueError: too many values to unpack
```

# PART B — Sets

## B1. What is a set?

### Concept

A **set** is an *unordered* collection of **unique** elements. Write the
same value twice - it is stored once. There are no positions, so there is
no indexing and no slicing.

### Why do we need it?

* Remove duplicates from a list in one call.
* Ask "is this value present?" — sets answer *fast* (Module 16 explains why).
* Compare groups: common courses, shared interests, missing skills.

### Syntax

```python
colours = {"red", "green", "blue"}
empty = set()          # NOT {} — that creates an empty dictionary!
```

````python
colours = {"red", "green", "blue"}
numbers = {1, 2, 2, 3, 3, 3}      # duplicates collapse automatically
empty = set()

print(numbers)
print(len(numbers))               # only 3 unique values
print(empty)
````

### Explanation

* `{1, 2, 2, 3, 3, 3}` becomes `{1, 2, 3}` — uniqueness is automatic.
* An **empty set** must be written `set()`; `{}` is an empty *dictionary*
  (Part C).
* Sets print in an arbitrary order — never rely on position.

Sets are also the fastest way to de-duplicate a list:

````python
votes = ["Ali", "Sara", "Ali", "Omar", "Sara", "Ali"]
unique_voters = set(votes)
print("Unique voters:", sorted(unique_voters))   # sorted() -> readable order
print("Distinct count:", len(unique_voters))
````

(We print `sorted(unique_voters)` because sets themselves have no order -
`sorted()` hands us an ordinary list in a predictable order.)

## B2. Adding and removing elements

````python
skills = {"python"}

skills.add("sql")            # add one element
print(sorted(skills))

skills.add("python")         # adding a duplicate: silently ignored
print(sorted(skills))

skills.discard("excel")      # remove if present; NO error if absent
skills.remove("sql")         # remove; raises KeyError if absent!
print(sorted(skills))

skills.pop()                 # removes SOME element (arbitrary for a set)
print("size after pop:", len(skills))
````

(The `pop()` line's output varies by machine — a set has no "first"
element to pop. That unpredictability is itself the lesson: do not use
`pop()` when you care *which* element leaves; use `remove()`/`discard()`.)

`clear()` empties a set completely — same as for lists:

````python
temp = {1, 2, 3}
temp.clear()
print(temp)
````

## B3. Membership testing — a set's killer feature

````python
allowed_ids = {101, 205, 309, 412}

print(205 in allowed_ids)          # fast True/False answer
print(999 in allowed_ids)
print(999 not in allowed_ids)
````

Sets answer membership questions *much* faster than lists when the
collection is large — Module 16 quantifies this. Rule of thumb: **looking
things up → set or dictionary; keeping order → list.**

## B4. Set operations

### Concept

| Operation | Operator | Method form | Meaning |
|---|---|---|---|
| union | `A \| B` | `A.union(B)` | elements in A **or** B (or both) |
| intersection | `A & B` | `A.intersection(B)` | elements in **both** |
| difference | `A - B` | `A.difference(B)` | in A but **not** in B |
| symmetric difference | `A ^ B` | `A.symmetric_difference(B)` | in exactly **one** of them |

````python
python_students = {"Ali", "Sara", "Omar", "Zainab"}
stats_students = {"Sara", "Omar", "Bilal"}

print("Union        :", sorted(python_students | stats_students))
print("Intersection :", sorted(python_students & stats_students))
print("Py - Stats   :", sorted(python_students - stats_students))
print("Stats - Py   :", sorted(stats_students - python_students))
print("Symmetric    :", sorted(python_students ^ stats_students))
````

(`sorted(...)` gives each result as a readable list — remember the set
itself has no order; the *content* is what matters.)

### Example — Practical: who takes exactly one of the two courses?

````python
python_students = {"Ali", "Sara", "Omar", "Zainab"}
stats_students = {"Sara", "Omar", "Bilal"}

exactly_one = python_students ^ stats_students
print("In exactly one course:", sorted(exactly_one))
````

## B5. Subset, superset, disjoint

````python
basic = {"print", "if", "for"}
python = {"print", "if", "for", "while", "def"}

print(basic <= python)            # subset: is basic inside python?
print(python >= basic)            # superset: does python contain basic?
print(basic.isdisjoint({"Java", "C"}))   # no common elements?
````

```python
# 4. Expecting a stable print order or sorted output:
#    print(sorted(skills))  # sorted() gives you a list if you need order
```

# PART C — Dictionaries

## C1. What is a dictionary?

### Concept

A **dictionary** stores **key → value** pairs. Instead of positions, you
look values up by a meaningful **key**. Real-world dictionaries map words
to definitions; Python dictionaries map keys to values.

### Why do we need it?

A list answers "what is at position 3?" A dictionary answers "what is
Ali's marks?" - self-describing data that survives reordering.

### Syntax

```python
student = {"name": "Ali", "age": 21, "program": "Data Science"}
student["age"]          # 21
```

````python
student = {"name": "Ali", "age": 21, "program": "Data Science"}

print(student)
print(student["name"])
print(student["age"])
````

### Valid keys

Keys must be *immutable*: strings and numbers are perfect; lists are not.
Values may be anything - including lists and other dictionaries.

````python

ok = {"name": "Ali", 42: "answer", 3.5: "half"}
print(ok[42], ok[3.5], ok["name"])

# The next line would fail — keys cannot be lists:
# bad = {[1, 2]: "oops"}      # TypeError: unhashable type: 'list'
````

````python
ok = {"name": "Ali", 42: "answer", 3.5: "half", ('a','b'): "tuple"}
print(ok[42], ok[3.5], ok["name"], ok[('a','b')])

# The next line would fail — keys cannot be lists:
# bad = {[1, 2]: "oops"}      # TypeError: unhashable type: 'list'
````

## C2. Accessing: `[]` vs `.get()`

### Concept

* `student["key"]` — **crashes** with `KeyError` if the key is missing.
* `student.get("key")` — returns `None` instead (or a default you give).

The same word, two behaviours: one strict, one forgiving.

````python
student = {"name": "Ali", "age": 21}

print(student["name"])           # present — both forms work
print(student.get("name"))

# The strict form would crash here (commented for the notebook's health):
# print(student["marks"])      # KeyError: 'marks'

print(student.get("marks"))            # missing -> None
print(student.get("marks", 0))         # missing -> your chosen default
````

**Which to use?** If a missing key means *your data is broken*, use `[]`
so the crash points at the bug. If a missing key is a *normal possibility*
("does the student have optional marks yet?"), use `.get()`.

## C3. Adding, updating, removing

````python
student = {"name": "Ali", "age": 21}

student["program"] = "Data Science"     # add a NEW key
student["age"] = 22                     # update an EXISTING key
print(student)

removed_age = student.pop("age")        # remove a key, get its value back
print("removed:", removed_age)
print(student)

last = student.popitem()                # removes the LAST inserted pair
print("popitem:", last)
print(student)
````

`update()` merges one dictionary into another — new keys are added,
existing keys are overwritten:

````python
defaults = {"theme": "light", "volume": 5}
user_prefs = {"volume": 8, "language": "en"}

defaults.update(user_prefs)
print(defaults)
````

## C4. Looping over a dictionary

### Concept

Three views: `keys()`, `values()`, and `items()` (pairs). Looping over the
dictionary itself loops over its keys. The most useful is `items()` with
tuple unpacking from Part A!

````python
marks = {"Ali": 78, "Sara": 91, "Omar": 84}

for name in marks.keys():                       # keys by default
    print(name, "->", marks[name])
````

````python
for name, score in marks.items():        # keys AND values together
    print(f"{name}: {score}")
````

````python
total = 0
for score in marks.values():             # just the values
    total += score
print("Class total:", total)
````

## C5. Dictionary membership - keys, not values

````python
student = {"name": "Ali", "age": 21}

print("name" in student)         # checks KEYS
print(21 in student)             # values are NOT checked by in
print(21 in student.values())    # ...unless you ask for the values view
````

## C6. Nested dictionaries — real records

Values can be dictionaries themselves. One student, fully described:

````python
student = {
    "name": "Amina",
    "roll_number": "CS-2104",
    "marks": {"math": 87, "stats": 91, "python": 84},
    "contact": {"email": "amina@uni.edu", "city": "Lahore"},
}

print(student["marks"]["stats"])        # two lookups, nested
print(student["contact"]["city"])

student["marks"]["python"] = 88         # update deep inside the nest
print(student["marks"])
````

### Example — Practical: frequency counting (the classic pattern)

````python
text = "data data science data python science , data data every where"

counts = {}
for word in text.split():
    if word in counts:
        counts[word] += 1          # seen before -> increase
    else:
        counts[word] = 1           # first time -> start at 1

print(counts)
````

Memorise this pattern - counting occurrences is half of data science, and
the plain-`if/else` version above is the one to know by heart.

### Common Mistakes (Part C)

```python
# 1. student["marks"]          # KeyError when absent — prefer .get() if optional
# 2. for k, v in student:      # ValueError — you must loop over .items()
# 3. Thinking `in` checks values — it checks keys.
# 4. Using a list as a key:    # {[1,2]: "x"}  -> TypeError (unhashable)
# 5. pop("missing") crashes; pop("missing", default) does not.
```

# D. Choosing the right container

| Structure | Ordered | Mutable | Duplicates | Main use |
|---|---|---|---|---|
| **List** | Yes | Yes | Yes | Ordered collections you will change |
| **Tuple** | Yes | **No** | Yes | Fixed records that must not change |
| **Set** | No guaranteed positional order | Yes | **No** | Uniqueness, membership, group maths |
| **Dictionary** | Insertion ordered (by key) | Yes | Keys unique | Key → value lookups |

How to decide:

* Need order + changes → **list**. Fixed record → **tuple**.
* Need "have I seen this?" or no duplicates → **set**.
* Need to look things up by a *name*, not a position → **dictionary**.

````python
# One cell, four containers, one fact each:
primes_list = [2, 3, 5, 7]           # ordered, changeable
origin_point = (0, 0)                # fixed coordinate
seen_ids = {101, 205}                # unique membership
population = {"Lahore": 13_000_000}  # key -> value lookup

primes_list.append(11)
print(primes_list, origin_point, sorted(seen_ids), population)
````

# E. Mini project — Contact Book

**Requirements**

1. Menu: add / search / list / delete / quit.
2. Contacts stored as `name -> phone` in one dictionary.
3. `search` uses `.get()` so unknown names are handled gracefully.
4. `delete` uses `in` before `pop()`.
5. Names are normalised (`.strip()`, `.title()`) so "ali" and "Ali" match.

**Pseudocode**

```text
contacts <- empty dictionary
LOOP forever:
    read choice
    IF add:     read name, phone; contacts[name] = phone
    IF search:  print contacts.get(name, "not found")
    IF list:    FOR name, phone IN contacts.items(): print pair
    IF delete:  IF name IN contacts: pop it; ELSE message
    IF quit:    break
```

````python
# ----- Contact Book -----
contacts = {}

while True:
    print()
    print("--- Contact Book ---")
    print("add / search / list / delete / quit")
    choice = input("Choice: ").strip().lower()

    if choice == "add":
        name = input("Name: ").strip().title()
        phone = input("Phone: ").strip()
        contacts[name] = phone
        print(f"Saved {name}: {phone}")
    elif choice == "search":
        name = input("Name to find: ").strip().title()
        phone = contacts.get(name)
        if phone is None:
            print(f"No contact named {name}.")
        else:
            print(f"{name}: {phone}")
    elif choice == "list":
        if len(contacts) == 0:
            print("The contact book is empty.")
        else:
            for name, phone in contacts.items():
                print(f"{name:<10} {phone}")
    elif choice == "delete":
        name = input("Name to delete: ").strip().title()
        if name in contacts:
            contacts.pop(name)
            print(f"Deleted {name}.")
        else:
            print(f"No contact named {name}.")
    elif choice == "quit":
        print("Goodbye! Contacts saved this session:", len(contacts))
        break
    else:
        print("Unknown choice.")
````

# H. Summary

* **Tuples** are immutable ordered records: create with commas, read like
  lists, unpack into variables, `count()`/`index()` are their only methods.
* **Sets** store unique values with no positions: `add`/`discard`/
  `remove`, blazing membership tests, and the four operations
  `| & - ^`.
* **Dictionaries** map keys to values: create, read with `[]` (strict) or
  `.get()` (forgiving), update by assignment, merge with `update()`,
  loop with `.items()`, count with the frequency pattern.
* `in` checks set membership and dictionary *keys*.
* Choose containers deliberately: list (ordered, changeable), tuple
  (fixed), set (unique), dict (keyed).
